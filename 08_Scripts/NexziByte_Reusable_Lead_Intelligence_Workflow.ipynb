```python
# ============================================================
# STEP 2 — AUTOMATIC RAW CSV DETECTION & LOADING
# ============================================================

# Find all CSV files in the Raw Data folder
raw_csv_files = [
    f for f in os.listdir(RAW_DIR)
    if f.lower().endswith(".csv")
]

print("CSV files found in 01_Raw_Data:")
print("-" * 60)

if not raw_csv_files:
    raise FileNotFoundError(
        "No CSV file found in 01_Raw_Data. "
        "Please place your raw B2B/Apollo CSV file in that folder."
    )

for i, filename in enumerate(raw_csv_files, start=1):
    print(f"{i}. {filename}")


# ------------------------------------------------------------
# SELECT RAW FILE
# ------------------------------------------------------------

# Automatically select the first CSV
RAW_FILE_NAME = raw_csv_files[0]
RAW_FILE_PATH = os.path.join(RAW_DIR, RAW_FILE_NAME)

print("\nSelected Raw File:")
print(RAW_FILE_NAME)


# ------------------------------------------------------------
# LOAD CSV
# ------------------------------------------------------------

try:
    raw_df = pd.read_csv(
        RAW_FILE_PATH,
        encoding="utf-8-sig",
        low_memory=False
    )
    
except UnicodeDecodeError:
    raw_df = pd.read_csv(
        RAW_FILE_PATH,
        encoding="latin1",
        low_memory=False
    )


# ------------------------------------------------------------
# BASIC INFORMATION
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("RAW DATA LOADED SUCCESSFULLY")
print("=" * 60)

print(f"File       : {RAW_FILE_NAME}")
print(f"Rows       : {raw_df.shape[0]}")
print(f"Columns    : {raw_df.shape[1]}")
print(f"Duplicates : {raw_df.duplicated().sum()}")
print(f"Missing    : {raw_df.isna().sum().sum()}")


# ------------------------------------------------------------
# COLUMN INVENTORY
# ------------------------------------------------------------

column_inventory = pd.DataFrame({
    "Column_Number": range(1, len(raw_df.columns) + 1),
    "Column_Name": raw_df.columns,
    "Data_Type": raw_df.dtypes.astype(str).values,
    "Missing_Count": raw_df.isna().sum().values,
    "Missing_Percentage": (
        raw_df.isna().mean().values * 100
    ).round(2),
    "Unique_Count": raw_df.nunique(dropna=True).values
})

print("\nColumn Inventory:")
display(column_inventory)


# ------------------------------------------------------------
# INITIAL DATA PREVIEW
# ------------------------------------------------------------

print("\nFirst 5 rows:")
display(raw_df.head())


# ------------------------------------------------------------
# SAVE INITIAL COLUMN INVENTORY
# ------------------------------------------------------------

column_inventory.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Initial_Column_Inventory.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

print(
    "\nInitial column inventory saved to 02_Data_Quality."
)

print("\nSTEP 2 COMPLETE.")
```

```python
# ============================================================
# STEP 3 — COLUMN NAME STANDARDIZATION + DATA TYPE NORMALIZATION
# ============================================================

import re

# ------------------------------------------------------------
# 1. KEEP ORIGINAL COLUMN NAMES FOR AUDIT
# ------------------------------------------------------------

original_columns = raw_df.columns.tolist()

# ------------------------------------------------------------
# 2. STANDARDIZE COLUMN NAMES
# ------------------------------------------------------------

def standardize_column_name(column):
    column = str(column).strip()
    
    # Replace line breaks/tabs with spaces
    column = re.sub(r"[\r\n\t]+", " ", column)
    
    # Remove extra spaces
    column = re.sub(r"\s+", " ", column)
    
    # Replace common separators
    column = column.replace("/", "_")
    column = column.replace("-", "_")
    
    # Remove special characters
    column = re.sub(r"[^A-Za-z0-9_ ]+", "", column)
    
    # Convert spaces to underscores
    column = column.replace(" ", "_")
    
    # Remove repeated underscores
    column = re.sub(r"_+", "_", column)
    
    # Remove leading/trailing underscores
    column = column.strip("_")
    
    return column


standardized_columns = [
    standardize_column_name(col)
    for col in raw_df.columns
]


# ------------------------------------------------------------
# 3. HANDLE DUPLICATE COLUMN NAMES
# ------------------------------------------------------------

def make_unique_columns(columns):
    seen = {}
    unique_columns = []
    
    for col in columns:
        if col not in seen:
            seen[col] = 0
            unique_columns.append(col)
        else:
            seen[col] += 1
            unique_columns.append(
                f"{col}_{seen[col]}"
            )
    
    return unique_columns


standardized_columns = make_unique_columns(
    standardized_columns
)

raw_df.columns = standardized_columns


# ------------------------------------------------------------
# 4. CREATE COLUMN MAPPING AUDIT
# ------------------------------------------------------------

column_mapping = pd.DataFrame({
    "Original_Column_Name": original_columns,
    "Standardized_Column_Name": standardized_columns
})

display(column_mapping)


# ------------------------------------------------------------
# 5. NORMALIZE TEXT VALUES
# ------------------------------------------------------------

text_columns = raw_df.select_dtypes(
    include=["object", "string"]
).columns

for col in text_columns:
    raw_df[col] = (
        raw_df[col]
        .astype("string")
        .str.strip()
        .replace({
            "": pd.NA,
            "nan": pd.NA,
            "None": pd.NA,
            "NULL": pd.NA,
            "N/A": pd.NA,
            "NA": pd.NA,
            "null": pd.NA
        })
    )


# ------------------------------------------------------------
# 6. NORMALIZE BOOLEAN-LIKE VALUES
# ------------------------------------------------------------

boolean_map = {
    "true": True,
    "false": False,
    "yes": True,
    "no": False,
    "y": True,
    "n": False
}

for col in raw_df.select_dtypes(
    include=["object", "string"]
).columns:
    
    lower_values = (
        raw_df[col]
        .astype("string")
        .str.lower()
    )
    
    unique_values = set(
        lower_values.dropna().unique()
    )
    
    if unique_values and unique_values.issubset(
        set(boolean_map.keys())
    ):
        raw_df[col] = lower_values.map(boolean_map)


# ------------------------------------------------------------
# 7. ATTEMPT NUMERIC CONVERSION FOR NUMERIC-LIKE COLUMNS
# ------------------------------------------------------------

for col in raw_df.columns:
    
    if raw_df[col].dtype == "object" or str(
        raw_df[col].dtype
    ) == "string":
        
        converted = pd.to_numeric(
            raw_df[col],
            errors="coerce"
        )
        
        non_null_original = raw_df[col].notna().sum()
        non_null_converted = converted.notna().sum()
        
        # Convert only when most non-null values are numeric
        if (
            non_null_original > 0
            and non_null_converted / non_null_original >= 0.90
        ):
            raw_df[col] = converted


# ------------------------------------------------------------
# 8. BASIC DATA TYPE SUMMARY
# ------------------------------------------------------------

dtype_summary = pd.DataFrame({
    "Column": raw_df.columns,
    "Data_Type": raw_df.dtypes.astype(str).values,
    "Non_Null": raw_df.notna().sum().values,
    "Null_Count": raw_df.isna().sum().values
})

display(dtype_summary)


# ------------------------------------------------------------
# 9. SAVE COLUMN MAPPING
# ------------------------------------------------------------

column_mapping.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Column_Mapping.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 10. SAVE INTERMEDIATE CLEAN STRUCTURE
# ------------------------------------------------------------

intermediate_path = os.path.join(
    CLEAN_DIR,
    "Reusable_Workflow_Step_03_Standardized.csv"
)

raw_df.to_csv(
    intermediate_path,
    index=False,
    encoding="utf-8-sig"
)


print("=" * 65)
print("STEP 3 COMPLETE")
print("=" * 65)

print("Columns standardized :", len(raw_df.columns))
print("Rows preserved        :", len(raw_df))
print("Column mapping saved  :", 
      "02_Data_Quality/Reusable_Workflow_Column_Mapping.csv")
print("Intermediate file     :", 
      "04_Clean_Data/Reusable_Workflow_Step_03_Standardized.csv")
```

```python
# ============================================================
# STEP 4 — DATA CLEANING & STANDARDIZATION
# ============================================================

# ------------------------------------------------------------
# 1. CREATE WORKING COPY
# ------------------------------------------------------------

clean_df = raw_df.copy()

print("Starting rows:", len(clean_df))
print("Starting columns:", len(clean_df.columns))


# ------------------------------------------------------------
# 2. HELPER FUNCTIONS
# ------------------------------------------------------------

def clean_text(value):
    if pd.isna(value):
        return pd.NA
    
    value = str(value).strip()
    value = re.sub(r"\s+", " ", value)
    
    if value.lower() in [
        "", "nan", "none", "null", "n/a", "na", "-"
    ]:
        return pd.NA
    
    return value


def clean_company_name(value):
    value = clean_text(value)
    
    if pd.isna(value):
        return pd.NA
    
    return value.strip()


def clean_url(value):
    value = clean_text(value)
    
    if pd.isna(value):
        return pd.NA
    
    value = value.strip()
    
    # Remove common accidental prefixes
    value = re.sub(
        r"^(https?://)?(www\.)?",
        "",
        value,
        flags=re.IGNORECASE
    )
    
    value = value.rstrip("/")
    
    return value.lower()


def clean_email(value):
    value = clean_text(value)
    
    if pd.isna(value):
        return pd.NA
    
    return value.lower().strip()


def clean_phone(value):
    value = clean_text(value)
    
    if pd.isna(value):
        return pd.NA
    
    # Preserve + and digits
    cleaned = re.sub(r"[^\d+]", "", value)
    
    return cleaned


# ------------------------------------------------------------
# 3. CLEAN ALL TEXT COLUMNS
# ------------------------------------------------------------

for col in clean_df.select_dtypes(
    include=["object", "string"]
).columns:
    
    clean_df[col] = clean_df[col].apply(clean_text)


# ------------------------------------------------------------
# 4. IDENTIFY IMPORTANT COLUMNS
# ------------------------------------------------------------

def find_column(possible_names):
    
    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in clean_df.columns
    }
    
    for name in possible_names:
        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )
        
        if key in normalized:
            return normalized[key]
    
    return None


company_col = find_column([
    "Company_Name",
    "Company",
    "CompanyName"
])

website_col = find_column([
    "Website",
    "Company_Website",
    "CompanyWebsite"
])

linkedin_col = find_column([
    "Company_Linkedin_Url",
    "Company_LinkedIn_Url",
    "LinkedIn_URL",
    "CompanyLinkedinUrl"
])

email_col = find_column([
    "Email",
    "Email_Address",
    "EmailAddress"
])

phone_col = find_column([
    "Phone",
    "Phone_Number",
    "PhoneNumber"
])

country_col = find_column([
    "Company_Country",
    "Country"
])

state_col = find_column([
    "Company_State",
    "State"
])

city_col = find_column([
    "Company_City",
    "City"
])

industry_col = find_column([
    "Industry",
    "Industry_Standardized"
])

print("\nDetected important columns:")
print("Company :", company_col)
print("Website :", website_col)
print("LinkedIn:", linkedin_col)
print("Email   :", email_col)
print("Phone   :", phone_col)
print("Country :", country_col)
print("State   :", state_col)
print("City    :", city_col)
print("Industry:", industry_col)


# ------------------------------------------------------------
# 5. STANDARDIZE COMPANY NAME
# ------------------------------------------------------------

if company_col:
    clean_df[company_col] = clean_df[
        company_col
    ].apply(clean_company_name)


# ------------------------------------------------------------
# 6. STANDARDIZE WEBSITE
# ------------------------------------------------------------

if website_col:
    clean_df[website_col] = clean_df[
        website_col
    ].apply(clean_url)


# ------------------------------------------------------------
# 7. STANDARDIZE LINKEDIN URL
# ------------------------------------------------------------

if linkedin_col:
    clean_df[linkedin_col] = clean_df[
        linkedin_col
    ].apply(clean_url)


# ------------------------------------------------------------
# 8. STANDARDIZE EMAIL
# ------------------------------------------------------------

if email_col:
    clean_df[email_col] = clean_df[
        email_col
    ].apply(clean_email)


# ------------------------------------------------------------
# 9. STANDARDIZE PHONE
# ------------------------------------------------------------

if phone_col:
    clean_df[phone_col] = clean_df[
        phone_col
    ].apply(clean_phone)


# ------------------------------------------------------------
# 10. STANDARDIZE LOCATION TEXT
# ------------------------------------------------------------

for col in [
    country_col,
    state_col,
    city_col
]:
    
    if col:
        clean_df[col] = (
            clean_df[col]
            .apply(clean_text)
        )


# ------------------------------------------------------------
# 11. STANDARDIZE INDUSTRY TEXT
# ------------------------------------------------------------

if industry_col:
    
    clean_df[industry_col] = (
        clean_df[industry_col]
        .apply(clean_text)
        .str.strip()
    )


# ------------------------------------------------------------
# 12. REMOVE COMPLETELY EMPTY ROWS
# ------------------------------------------------------------

before_empty_removal = len(clean_df)

clean_df = clean_df.dropna(
    how="all"
).reset_index(drop=True)

empty_rows_removed = (
    before_empty_removal - len(clean_df)
)


# ------------------------------------------------------------
# 13. CREATE CLEANING REPORT
# ------------------------------------------------------------

cleaning_report = pd.DataFrame({
    "Metric": [
        "Rows Before Cleaning",
        "Rows After Cleaning",
        "Completely Empty Rows Removed",
        "Columns",
        "Duplicate Rows Remaining",
        "Missing Cells Remaining"
    ],
    "Value": [
        len(raw_df),
        len(clean_df),
        empty_rows_removed,
        len(clean_df.columns),
        clean_df.duplicated().sum(),
        clean_df.isna().sum().sum()
    ]
})

display(cleaning_report)


# ------------------------------------------------------------
# 14. SAVE CLEANED DATA
# ------------------------------------------------------------

clean_output_path = os.path.join(
    CLEAN_DIR,
    "Reusable_Workflow_Cleaned_Data.csv"
)

clean_df.to_csv(
    clean_output_path,
    index=False,
    encoding="utf-8-sig"
)

cleaning_report.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Cleaning_Report.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. FINAL CHECK
# ------------------------------------------------------------

print("=" * 65)
print("STEP 4 COMPLETE")
print("=" * 65)

print("Rows before cleaning :", len(raw_df))
print("Rows after cleaning  :", len(clean_df))
print("Rows removed         :", len(raw_df) - len(clean_df))
print("Columns preserved    :", len(clean_df.columns))

print("\nCleaned data saved:")
print(clean_output_path)
```

```python
# ============================================================
# STEP 5 — DUPLICATE DETECTION & DECISION LOGIC
# ============================================================

# ------------------------------------------------------------
# 1. START FROM CLEAN DATA
# ------------------------------------------------------------

duplicate_df = clean_df.copy()

print("Rows entering duplicate detection:", len(duplicate_df))


# ------------------------------------------------------------
# 2. HELPER — FIND COLUMN
# ------------------------------------------------------------

def find_existing_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


dup_company_col = find_existing_column(
    duplicate_df,
    ["Company_Name", "Company", "CompanyName"]
)

dup_email_col = find_existing_column(
    duplicate_df,
    ["Email", "Email_Address", "EmailAddress"]
)

dup_website_col = find_existing_column(
    duplicate_df,
    ["Website", "Company_Website", "CompanyWebsite"]
)

dup_linkedin_col = find_existing_column(
    duplicate_df,
    [
        "Company_Linkedin_Url",
        "Company_LinkedIn_Url",
        "LinkedIn_URL",
        "CompanyLinkedinUrl"
    ]
)

dup_country_col = find_existing_column(
    duplicate_df,
    ["Company_Country", "Country"]
)

dup_city_col = find_existing_column(
    duplicate_df,
    ["Company_City", "City"]
)


print("\nDuplicate detection columns:")
print("Company :", dup_company_col)
print("Email   :", dup_email_col)
print("Website :", dup_website_col)
print("LinkedIn:", dup_linkedin_col)
print("Country :", dup_country_col)
print("City    :", dup_city_col)


# ------------------------------------------------------------
# 3. EXACT DUPLICATE ROWS
# ------------------------------------------------------------

duplicate_df["Exact_Duplicate"] = (
    duplicate_df.duplicated(
        keep=False
    )
)


# ------------------------------------------------------------
# 4. NORMALIZED IDENTIFIERS
# ------------------------------------------------------------

def normalize_identifier(value):

    if pd.isna(value):
        return pd.NA

    value = str(value).lower().strip()

    # Remove protocol
    value = re.sub(
        r"^https?://",
        "",
        value
    )

    # Remove www
    value = re.sub(
        r"^www\.",
        "",
        value
    )

    # Remove trailing slash
    value = value.rstrip("/")

    # Remove spaces
    value = re.sub(
        r"\s+",
        "",
        value
    )

    return value


if dup_email_col:
    duplicate_df["_Normalized_Email"] = (
        duplicate_df[dup_email_col]
        .apply(normalize_identifier)
    )

if dup_website_col:
    duplicate_df["_Normalized_Website"] = (
        duplicate_df[dup_website_col]
        .apply(normalize_identifier)
    )

if dup_linkedin_col:
    duplicate_df["_Normalized_LinkedIn"] = (
        duplicate_df[dup_linkedin_col]
        .apply(normalize_identifier)
    )

if dup_company_col:
    duplicate_df["_Normalized_Company"] = (
        duplicate_df[dup_company_col]
        .apply(normalize_identifier)
    )

if dup_country_col:
    duplicate_df["_Normalized_Country"] = (
        duplicate_df[dup_country_col]
        .apply(normalize_identifier)
    )

if dup_city_col:
    duplicate_df["_Normalized_City"] = (
        duplicate_df[dup_city_col]
        .apply(normalize_identifier)
    )


# ------------------------------------------------------------
# 5. CREATE DUPLICATE FLAGS
# ------------------------------------------------------------

duplicate_df["Duplicate_Type"] = "Unique"


# Email duplicate
if dup_email_col:

    mask = (
        duplicate_df["_Normalized_Email"].notna()
        &
        duplicate_df["_Normalized_Email"].duplicated(
            keep=False
        )
    )

    duplicate_df.loc[
        mask,
        "Duplicate_Type"
    ] = "Duplicate Email"


# Website duplicate
if dup_website_col:

    mask = (
        duplicate_df["_Normalized_Website"].notna()
        &
        duplicate_df["_Normalized_Website"].duplicated(
            keep=False
        )
    )

    duplicate_df.loc[
        mask
        &
        (duplicate_df["Duplicate_Type"] == "Unique"),
        "Duplicate_Type"
    ] = "Duplicate Website"


# LinkedIn duplicate
if dup_linkedin_col:

    mask = (
        duplicate_df["_Normalized_LinkedIn"].notna()
        &
        duplicate_df["_Normalized_LinkedIn"].duplicated(
            keep=False
        )
    )

    duplicate_df.loc[
        mask
        &
        (duplicate_df["Duplicate_Type"] == "Unique"),
        "Duplicate_Type"
    ] = "Duplicate LinkedIn"


# Company + location duplicate
if dup_company_col:

    if dup_country_col and dup_city_col:

        duplicate_df["_Company_Location_Key"] = (
            duplicate_df["_Normalized_Company"].fillna("")
            + "|"
            + duplicate_df["_Normalized_Country"].fillna("")
            + "|"
            + duplicate_df["_Normalized_City"].fillna("")
        )

    else:

        duplicate_df["_Company_Location_Key"] = (
            duplicate_df["_Normalized_Company"].fillna("")
        )


    mask = (
        duplicate_df["_Normalized_Company"].notna()
        &
        duplicate_df["_Company_Location_Key"].duplicated(
            keep=False
        )
    )

    duplicate_df.loc[
        mask
        &
        (duplicate_df["Duplicate_Type"] == "Unique"),
        "Duplicate_Type"
    ] = "Possible Company Duplicate"


# Exact duplicate gets highest priority
duplicate_df.loc[
    duplicate_df["Exact_Duplicate"],
    "Duplicate_Type"
] = "Exact Duplicate"


# ------------------------------------------------------------
# 6. DUPLICATE DECISION
# ------------------------------------------------------------

duplicate_df["Duplicate_Decision"] = "KEEP"


# Exact duplicates → remove later
duplicate_df.loc[
    duplicate_df["Duplicate_Type"] == "Exact Duplicate",
    "Duplicate_Decision"
] = "REMOVE"


# Strong identifier duplicates → review
strong_duplicate_types = [
    "Duplicate Email",
    "Duplicate Website",
    "Duplicate LinkedIn"
]

duplicate_df.loc[
    duplicate_df["Duplicate_Type"].isin(
        strong_duplicate_types
    ),
    "Duplicate_Decision"
] = "REVIEW"


# Similar company/location → review
duplicate_df.loc[
    duplicate_df["Duplicate_Type"]
    == "Possible Company Duplicate",
    "Duplicate_Decision"
] = "REVIEW"


# ------------------------------------------------------------
# 7. DUPLICATE SUMMARY
# ------------------------------------------------------------

duplicate_summary = (
    duplicate_df[
        "Duplicate_Type"
    ]
    .value_counts(dropna=False)
    .rename_axis("Duplicate_Type")
    .reset_index(name="Count")
)

print("\nDuplicate Summary:")
display(duplicate_summary)


# ------------------------------------------------------------
# 8. DECISION SUMMARY
# ------------------------------------------------------------

decision_summary = (
    duplicate_df[
        "Duplicate_Decision"
    ]
    .value_counts(dropna=False)
    .rename_axis("Duplicate_Decision")
    .reset_index(name="Count")
)

print("\nDuplicate Decision Summary:")
display(decision_summary)


# ------------------------------------------------------------
# 9. SHOW RECORDS REQUIRING REVIEW
# ------------------------------------------------------------

review_duplicates = duplicate_df[
    duplicate_df["Duplicate_Decision"] == "REVIEW"
].copy()

print(
    "\nRecords requiring duplicate review:",
    len(review_duplicates)
)

if len(review_duplicates) > 0:

    review_columns = []

    for col in [
        dup_company_col,
        dup_email_col,
        dup_website_col,
        dup_linkedin_col,
        dup_country_col,
        dup_city_col
    ]:

        if col and col not in review_columns:
            review_columns.append(col)

    review_columns += [
        "Duplicate_Type",
        "Duplicate_Decision"
    ]

    display(
        review_duplicates[
            review_columns
        ].head(50)
    )


# ------------------------------------------------------------
# 10. REMOVE ONLY EXACT DUPLICATES AUTOMATICALLY
# ------------------------------------------------------------

before_removal = len(duplicate_df)

deduplicated_df = duplicate_df[
    duplicate_df["Duplicate_Decision"] != "REMOVE"
].copy()

deduplicated_df = (
    deduplicated_df
    .drop_duplicates()
    .reset_index(drop=True)
)

removed_count = (
    before_removal - len(deduplicated_df)
)


# ------------------------------------------------------------
# 11. CREATE DATA QUALITY STATUS
# ------------------------------------------------------------

deduplicated_df["Data_Quality_Status"] = "Normal"

deduplicated_df.loc[
    deduplicated_df["Duplicate_Decision"] == "REVIEW",
    "Data_Quality_Status"
] = "Needs Verification"


# ------------------------------------------------------------
# 12. SAVE DUPLICATE REPORT
# ------------------------------------------------------------

duplicate_report_path = os.path.join(
    QUALITY_DIR,
    "Reusable_Workflow_Duplicate_Decision_Report.csv"
)

duplicate_df.to_csv(
    duplicate_report_path,
    index=False,
    encoding="utf-8-sig"
)


duplicate_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Duplicate_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 13. SAVE DEDUPLICATED DATA
# ------------------------------------------------------------

deduplicated_path = os.path.join(
    CLEAN_DIR,
    "Reusable_Workflow_Deduplicated_Data.csv"
)

deduplicated_df.to_csv(
    deduplicated_path,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 14. FINAL REPORT
# ------------------------------------------------------------

print("=" * 65)
print("STEP 5 COMPLETE")
print("=" * 65)

print("Rows before duplicate handling :", before_removal)
print("Exact duplicates removed       :", removed_count)
print("Rows after automatic removal   :", len(deduplicated_df))
print("Records requiring review       :", len(review_duplicates))

print("\nDuplicate report:")
print(duplicate_report_path)

print("\nDeduplicated data:")
print(deduplicated_path)
```

```python
# ============================================================
# STEP 6 — MISSING & INVALID DATA VALIDATION
# ============================================================

# Start from the deduplicated dataset
validation_df = deduplicated_df.copy()

print("Rows entering validation:", len(validation_df))


# ------------------------------------------------------------
# 1. FIND IMPORTANT COLUMNS
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


company_col = find_column(
    validation_df,
    ["Company_Name", "Company", "CompanyName"]
)

website_col = find_column(
    validation_df,
    ["Website", "Company_Website", "CompanyWebsite"]
)

email_col = find_column(
    validation_df,
    ["Email", "Email_Address", "EmailAddress"]
)

phone_col = find_column(
    validation_df,
    ["Phone", "Phone_Number", "PhoneNumber"]
)

industry_col = find_column(
    validation_df,
    ["Industry", "Industry_Standardized"]
)

country_col = find_column(
    validation_df,
    ["Company_Country", "Country"]
)

city_col = find_column(
    validation_df,
    ["Company_City", "City"]
)

employees_col = find_column(
    validation_df,
    [
        "#_Employees",
        "Employees",
        "Employee_Count",
        "Number_of_Employees"
    ]
)

revenue_col = find_column(
    validation_df,
    [
        "Annual_Revenue",
        "Revenue",
        "AnnualRevenue"
    ]
)


# ------------------------------------------------------------
# 2. CREATE VALIDATION FLAGS
# ------------------------------------------------------------

validation_df["Company_Name_Valid"] = True
validation_df["Website_Valid"] = True
validation_df["Email_Valid"] = True
validation_df["Phone_Valid"] = True
validation_df["Industry_Available"] = True
validation_df["Location_Available"] = True
validation_df["Employee_Data_Available"] = True
validation_df["Revenue_Data_Available"] = True


# ------------------------------------------------------------
# 3. COMPANY NAME VALIDATION
# ------------------------------------------------------------

if company_col:

    validation_df["Company_Name_Valid"] = (
        validation_df[company_col]
        .notna()
        &
        (
            validation_df[company_col]
            .astype(str)
            .str.len()
            >= 2
        )
    )


# ------------------------------------------------------------
# 4. WEBSITE VALIDATION
# ------------------------------------------------------------

if website_col:

    def valid_website(value):

        if pd.isna(value):
            return False

        value = str(value).strip().lower()

        return (
            "." in value
            and " " not in value
            and len(value) >= 4
        )

    validation_df["Website_Valid"] = (
        validation_df[website_col]
        .apply(valid_website)
    )


# ------------------------------------------------------------
# 5. EMAIL VALIDATION
# ------------------------------------------------------------

if email_col:

    email_pattern = (
        r"^[A-Za-z0-9._%+-]+@"
        r"[A-Za-z0-9.-]+\."
        r"[A-Za-z]{2,}$"
    )

    validation_df["Email_Valid"] = (
        validation_df[email_col]
        .fillna("")
        .astype(str)
        .str.match(
            email_pattern,
            na=False
        )
    )


# ------------------------------------------------------------
# 6. PHONE VALIDATION
# ------------------------------------------------------------

if phone_col:

    validation_df["Phone_Valid"] = (
        validation_df[phone_col]
        .fillna("")
        .astype(str)
        .str.replace(
            r"\D",
            "",
            regex=True
        )
        .str.len()
        .between(7, 15)
    )


# ------------------------------------------------------------
# 7. INDUSTRY AVAILABILITY
# ------------------------------------------------------------

if industry_col:

    validation_df["Industry_Available"] = (
        validation_df[industry_col]
        .notna()
    )


# ------------------------------------------------------------
# 8. LOCATION AVAILABILITY
# ------------------------------------------------------------

location_checks = []

if country_col:
    location_checks.append(
        validation_df[country_col].notna()
    )

if city_col:
    location_checks.append(
        validation_df[city_col].notna()
    )

if location_checks:

    validation_df["Location_Available"] = (
        pd.concat(
            location_checks,
            axis=1
        ).any(axis=1)
    )


# ------------------------------------------------------------
# 9. EMPLOYEE DATA
# ------------------------------------------------------------

if employees_col:

    validation_df["Employee_Data_Available"] = (
        validation_df[employees_col]
        .notna()
    )


# ------------------------------------------------------------
# 10. REVENUE DATA
# ------------------------------------------------------------

if revenue_col:

    validation_df["Revenue_Data_Available"] = (
        validation_df[revenue_col]
        .notna()
    )


# ------------------------------------------------------------
# 11. MISSING FIELD COUNT
# ------------------------------------------------------------

validation_flags = [
    "Company_Name_Valid",
    "Website_Valid",
    "Email_Valid",
    "Phone_Valid",
    "Industry_Available",
    "Location_Available",
    "Employee_Data_Available",
    "Revenue_Data_Available"
]

validation_df["Missing_or_Invalid_Field_Count"] = (
    (~validation_df[validation_flags])
    .sum(axis=1)
)


# ------------------------------------------------------------
# 12. DATA QUALITY STATUS
# ------------------------------------------------------------

def assign_quality_status(row):

    # Company name is fundamental
    if not row["Company_Name_Valid"]:
        return "Invalid / Do Not Use"

    # Multiple fundamental issues
    if row["Missing_or_Invalid_Field_Count"] >= 4:
        return "Needs Verification"

    # Some information missing/invalid
    if row["Missing_or_Invalid_Field_Count"] > 0:
        return "Needs Verification"

    return "Normal"


validation_df["Data_Quality_Status"] = (
    validation_df.apply(
        assign_quality_status,
        axis=1
    )
)


# ------------------------------------------------------------
# 13. VERIFICATION REASON
# ------------------------------------------------------------

def verification_reason(row):

    reasons = []

    if not row["Company_Name_Valid"]:
        reasons.append("Missing/Invalid Company Name")

    if not row["Website_Valid"]:
        reasons.append("Missing/Invalid Website")

    if not row["Email_Valid"]:
        reasons.append("Missing/Invalid Email")

    if not row["Phone_Valid"]:
        reasons.append("Missing/Invalid Phone")

    if not row["Industry_Available"]:
        reasons.append("Industry Missing")

    if not row["Location_Available"]:
        reasons.append("Location Missing")

    if not row["Employee_Data_Available"]:
        reasons.append("Employee Data Missing")

    if not row["Revenue_Data_Available"]:
        reasons.append("Revenue Data Missing")

    if not reasons:
        return "No immediate verification issue"

    return "; ".join(reasons)


validation_df["Verification_Reason"] = (
    validation_df.apply(
        verification_reason,
        axis=1
    )
)


# ------------------------------------------------------------
# 14. VALIDATION SUMMARY
# ------------------------------------------------------------

quality_summary = (
    validation_df["Data_Quality_Status"]
    .value_counts()
    .rename_axis("Data_Quality_Status")
    .reset_index(name="Count")
)

print("\nData Quality Summary:")
display(quality_summary)


# ------------------------------------------------------------
# 15. FIELD VALIDATION SUMMARY
# ------------------------------------------------------------

field_validation_summary = pd.DataFrame({
    "Field_Check": validation_flags,
    "Valid_or_Available": [
        validation_df[col].sum()
        for col in validation_flags
    ],
    "Missing_or_Invalid": [
        (~validation_df[col]).sum()
        for col in validation_flags
    ]
})

field_validation_summary[
    "Missing_or_Invalid_Percentage"
] = (
    field_validation_summary["Missing_or_Invalid"]
    / len(validation_df)
    * 100
).round(2)

print("\nField Validation Summary:")
display(field_validation_summary)


# ------------------------------------------------------------
# 16. SHOW RECORDS REQUIRING VERIFICATION
# ------------------------------------------------------------

verification_df = validation_df[
    validation_df["Data_Quality_Status"]
    != "Normal"
].copy()

print(
    "\nRecords requiring verification:",
    len(verification_df)
)

if len(verification_df) > 0:

    display_columns = []

    for col in [
        company_col,
        website_col,
        email_col,
        phone_col,
        industry_col,
        country_col,
        city_col
    ]:

        if col and col not in display_columns:
            display_columns.append(col)

    display_columns += [
        "Data_Quality_Status",
        "Missing_or_Invalid_Field_Count",
        "Verification_Reason"
    ]

    display(
        verification_df[
            display_columns
        ].head(50)
    )


# ------------------------------------------------------------
# 17. SAVE REPORTS
# ------------------------------------------------------------

quality_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Data_Quality_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

field_validation_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Field_Validation_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

verification_df.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Verification_Required.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 18. SAVE VALIDATED DATA
# ------------------------------------------------------------

validation_output = os.path.join(
    CLEAN_DIR,
    "Reusable_Workflow_Validated_Data.csv"
)

validation_df.to_csv(
    validation_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 6 COMPLETE")
print("=" * 65)

print("Rows validated :", len(validation_df))
print("\nData Quality:")
print(quality_summary.to_string(index=False))

print("\nValidated data saved:")
print(validation_output)
```

```python
# ============================================================
# STEP 7 — INDUSTRY, LOCATION & COMPANY-SIZE STANDARDIZATION
# ============================================================

enrichment_df = validation_df.copy()

print("Rows entering Step 7:", len(enrichment_df))


# ------------------------------------------------------------
# 1. HELPER TO FIND COLUMNS
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


# ------------------------------------------------------------
# 2. IDENTIFY COMPANY FIELDS
# ------------------------------------------------------------

industry_col = find_column(
    enrichment_df,
    [
        "Industry",
        "Industry_Standardized"
    ]
)

country_col = find_column(
    enrichment_df,
    [
        "Company_Country",
        "Country",
        "Country_Standardized"
    ]
)

state_col = find_column(
    enrichment_df,
    [
        "Company_State",
        "State"
    ]
)

city_col = find_column(
    enrichment_df,
    [
        "Company_City",
        "City"
    ]
)

employees_col = find_column(
    enrichment_df,
    [
        "#_Employees",
        "Employees",
        "Employee_Count",
        "Number_of_Employees",
        "#_Employees_"
    ]
)

revenue_col = find_column(
    enrichment_df,
    [
        "Annual_Revenue",
        "Revenue",
        "AnnualRevenue"
    ]
)

founded_col = find_column(
    enrichment_df,
    [
        "Founded_Year",
        "FoundedYear",
        "Year_Founded"
    ]
)


print("\nDetected fields:")
print("Industry :", industry_col)
print("Country  :", country_col)
print("State    :", state_col)
print("City     :", city_col)
print("Employees:", employees_col)
print("Revenue  :", revenue_col)
print("Founded  :", founded_col)


# ------------------------------------------------------------
# 3. INDUSTRY STANDARDIZATION
# ------------------------------------------------------------

if industry_col:

    enrichment_df["Industry_Standardized_Workflow"] = (
        enrichment_df[industry_col]
        .astype("string")
        .str.strip()
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
    )

    # Common formatting normalization
    enrichment_df[
        "Industry_Standardized_Workflow"
    ] = (
        enrichment_df[
            "Industry_Standardized_Workflow"
        ]
        .str.title()
    )

else:

    enrichment_df[
        "Industry_Standardized_Workflow"
    ] = "Not Available"


# ------------------------------------------------------------
# 4. INDUSTRY REVIEW FLAG
# ------------------------------------------------------------

def industry_review(value):

    if pd.isna(value):
        return "Needs Verification"

    value = str(value).strip()

    if value == "" or value.lower() == "not available":
        return "Needs Verification"

    return "Normal"


enrichment_df["Industry_Review_Flag_Workflow"] = (
    enrichment_df[
        "Industry_Standardized_Workflow"
    ].apply(industry_review)
)


# ------------------------------------------------------------
# 5. COUNTRY STANDARDIZATION
# ------------------------------------------------------------

country_map = {
    "usa": "United States",
    "us": "United States",
    "u.s.": "United States",
    "u.s.a.": "United States",
    "united states of america": "United States",
    "uk": "United Kingdom",
    "u.k.": "United Kingdom",
    "uae": "United Arab Emirates",
    "u.a.e.": "United Arab Emirates",
    "india": "India"
}


if country_col:

    enrichment_df[
        "Country_Standardized_Workflow"
    ] = (
        enrichment_df[country_col]
        .astype("string")
        .str.strip()
        .str.lower()
        .replace(country_map)
    )

    # Proper display formatting for values not in map
    enrichment_df[
        "Country_Standardized_Workflow"
    ] = (
        enrichment_df[
            "Country_Standardized_Workflow"
        ]
        .apply(
            lambda x:
            x.title()
            if pd.notna(x)
            and str(x).strip() != ""
            else pd.NA
        )
    )

else:

    enrichment_df[
        "Country_Standardized_Workflow"
    ] = pd.NA


# ------------------------------------------------------------
# 6. LOCATION STATUS
# ------------------------------------------------------------

enrichment_df[
    "Location_Data_Status_Workflow"
] = np.where(
    enrichment_df[
        "Country_Standardized_Workflow"
    ].notna(),
    "Available",
    "Needs Verification"
)


# ------------------------------------------------------------
# 7. CITY STANDARDIZATION
# ------------------------------------------------------------

if city_col:

    enrichment_df["City_Standardized_Workflow"] = (
        enrichment_df[city_col]
        .astype("string")
        .str.strip()
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
        .str.title()
    )

else:

    enrichment_df[
        "City_Standardized_Workflow"
    ] = pd.NA


# ------------------------------------------------------------
# 8. STATE STANDARDIZATION
# ------------------------------------------------------------

if state_col:

    enrichment_df["State_Standardized_Workflow"] = (
        enrichment_df[state_col]
        .astype("string")
        .str.strip()
        .str.replace(
            r"\s+",
            " ",
            regex=True
        )
        .str.title()
    )

else:

    enrichment_df[
        "State_Standardized_Workflow"
    ] = pd.NA


# ------------------------------------------------------------
# 9. EMPLOYEE DATA STATUS
# ------------------------------------------------------------

if employees_col:

    enrichment_df[employees_col] = pd.to_numeric(
        enrichment_df[employees_col],
        errors="coerce"
    )

    enrichment_df[
        "Employee_Data_Status_Workflow"
    ] = np.where(
        enrichment_df[employees_col].notna(),
        "Available",
        "Needs Verification"
    )

else:

    enrichment_df[
        "Employee_Data_Status_Workflow"
    ] = "Needs Verification"


# ------------------------------------------------------------
# 10. COMPANY SIZE CATEGORY
# ------------------------------------------------------------

if employees_col:

    def company_size_category(employees):

        if pd.isna(employees):
            return "Needs Verification"

        if employees <= 50:
            return "Startup/Small"

        elif employees <= 250:
            return "SME"

        elif employees <= 1000:
            return "Mid-Market"

        elif employees <= 5000:
            return "Large Enterprise"

        else:
            return "Enterprise"


    enrichment_df[
        "Company_Size_Category_Workflow"
    ] = enrichment_df[
        employees_col
    ].apply(company_size_category)

else:

    enrichment_df[
        "Company_Size_Category_Workflow"
    ] = "Needs Verification"


# ------------------------------------------------------------
# 11. REVENUE BAND
# ------------------------------------------------------------

if revenue_col:

    enrichment_df[revenue_col] = pd.to_numeric(
        enrichment_df[revenue_col],
        errors="coerce"
    )

    def revenue_band(revenue):

        if pd.isna(revenue):
            return "Needs Verification"

        if revenue < 1_000_000:
            return "< $1M"

        elif revenue < 10_000_000:
            return "$1M–$10M"

        elif revenue < 50_000_000:
            return "$10M–$50M"

        elif revenue < 250_000_000:
            return "$50M–$250M"

        else:
            return "$250M+"


    enrichment_df[
        "Revenue_Band_Workflow"
    ] = enrichment_df[
        revenue_col
    ].apply(revenue_band)

else:

    enrichment_df[
        "Revenue_Band_Workflow"
    ] = "Needs Verification"


# ------------------------------------------------------------
# 12. COMPANY AGE
# ------------------------------------------------------------

if founded_col:

    enrichment_df[founded_col] = pd.to_numeric(
        enrichment_df[founded_col],
        errors="coerce"
    )

    current_year = pd.Timestamp.now().year

    enrichment_df[
        "Company_Age_Workflow"
    ] = (
        current_year
        - enrichment_df[founded_col]
    )

    # Prevent impossible ages
    enrichment_df.loc[
        (
            enrichment_df["Company_Age_Workflow"] < 0
        )
        |
        (
            enrichment_df["Company_Age_Workflow"] > 300
        ),
        "Company_Age_Workflow"
    ] = np.nan

else:

    enrichment_df[
        "Company_Age_Workflow"
    ] = np.nan


# ------------------------------------------------------------
# 13. B2B / B2C CLASSIFICATION
# ------------------------------------------------------------

# Only classify when the raw data provides usable evidence.
# Otherwise use Needs Verification.

b2b_col = find_column(
    enrichment_df,
    [
        "Business_Model",
        "B2B_B2C",
        "B2B_B2C_Type",
        "Customer_Type"
    ]
)

if b2b_col:

    def normalize_business_model(value):

        if pd.isna(value):
            return "Needs Verification"

        value = str(value).strip().lower()

        if value in ["b2b", "business to business"]:
            return "B2B"

        if value in ["b2c", "business to consumer"]:
            return "B2C"

        if value in [
            "b2b/b2c",
            "b2b & b2c",
            "both"
        ]:
            return "B2B/B2C"

        return "Needs Verification"


    enrichment_df[
        "Business_Model_Workflow"
    ] = enrichment_df[
        b2b_col
    ].apply(normalize_business_model)

else:

    enrichment_df[
        "Business_Model_Workflow"
    ] = "Needs Verification"


# ------------------------------------------------------------
# 14. SUMMARY
# ------------------------------------------------------------

print("\nIndustry distribution:")
display(
    enrichment_df[
        "Industry_Standardized_Workflow"
    ]
    .value_counts()
    .head(20)
    .rename_axis("Industry")
    .reset_index(name="Count")
)


print("\nCompany Size distribution:")
display(
    enrichment_df[
        "Company_Size_Category_Workflow"
    ]
    .value_counts()
    .rename_axis("Company_Size")
    .reset_index(name="Count")
)


print("\nCountry distribution:")
display(
    enrichment_df[
        "Country_Standardized_Workflow"
    ]
    .value_counts()
    .head(20)
    .rename_axis("Country")
    .reset_index(name="Count")
)


# ------------------------------------------------------------
# 15. SAVE ENRICHMENT STANDARDIZATION REPORT
# ------------------------------------------------------------

standardization_summary = pd.DataFrame({
    "Metric": [
        "Rows",
        "Columns",
        "Industries",
        "Countries",
        "Company Size Categories",
        "B2B/B2C Available"
    ],
    "Value": [
        len(enrichment_df),
        len(enrichment_df.columns),
        enrichment_df[
            "Industry_Standardized_Workflow"
        ].nunique(dropna=True),
        enrichment_df[
            "Country_Standardized_Workflow"
        ].nunique(dropna=True),
        enrichment_df[
            "Company_Size_Category_Workflow"
        ].nunique(dropna=True),
        (
            enrichment_df[
                "Business_Model_Workflow"
            ] != "Needs Verification"
        ).sum()
    ]
})

standardization_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Standardization_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 16. SAVE STEP OUTPUT
# ------------------------------------------------------------

step7_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_07_Standardized_Company_Data.csv"
)

enrichment_df.to_csv(
    step7_output,
    index=False,
    encoding="utf-8-sig"
)


print("=" * 65)
print("STEP 7 COMPLETE")
print("=" * 65)
print("Rows preserved:", len(enrichment_df))
print("Output:", step7_output)
```

```python
# ============================================================
# STEP 8 — TECHNOLOGY & BUSINESS SIGNAL PROCESSING
# ============================================================

signal_df = enrichment_df.copy()

print("Rows entering Step 8:", len(signal_df))


# ------------------------------------------------------------
# 1. HELPER FUNCTIONS
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


def signal_value(value):

    if pd.isna(value):
        return "Not Available"

    value = str(value).strip()

    if value == "":
        return "Not Available"

    if value.lower() in [
        "nan",
        "none",
        "null",
        "n/a",
        "na",
        "not available"
    ]:
        return "Not Available"

    return value


# ------------------------------------------------------------
# 2. LOCATE EXISTING SIGNAL COLUMNS
# ------------------------------------------------------------

signal_column_map = {

    "AI_ML_Signals": [
        "AI_ML_Signals",
        "AI_ML_Signal",
        "AI_ML"
    ],

    "Cloud_Data_Signals": [
        "Cloud_Data_Signals",
        "Cloud_Data_Signal",
        "Cloud_Data"
    ],

    "API_Platform_Signals": [
        "API_Platform_Signals",
        "API_Platform_Signal",
        "API_Platform"
    ],

    "Software_Development_Signals": [
        "Software_Development_Signals",
        "Software_Development_Signal",
        "Software_Development"
    ],

    "Hiring_Growth_Signals": [
        "Hiring_Growth_Signals",
        "Hiring_Growth_Signal",
        "Hiring_Growth"
    ],

    "Funding_Expansion_Signals": [
        "Funding_Expansion_Signals",
        "Funding_Expansion_Signal",
        "Funding_Expansion"
    ],

    "Automation_Workflow_Signals": [
        "Automation_Workflow_Signals",
        "Automation_Workflow_Signal",
        "Automation_Signals"
    ],

    "Product_Launch_Signals": [
        "Product_Launch_Signals",
        "Product_Launch_Signal"
    ],

    "Partnership_Signals": [
        "Partnership_Signals",
        "Partnership_Signal"
    ]
}


detected_signal_columns = {}

for standard_name, possible_names in signal_column_map.items():

    detected_signal_columns[standard_name] = find_column(
        signal_df,
        possible_names
    )

print("\nDetected signal columns:")

for standard_name, actual_column in detected_signal_columns.items():
    print(
        f"{standard_name:<35} → {actual_column}"
    )


# ------------------------------------------------------------
# 3. CREATE STANDARD SIGNAL FIELDS
# ------------------------------------------------------------

for standard_name, actual_column in detected_signal_columns.items():

    if actual_column:

        signal_df[standard_name] = (
            signal_df[actual_column]
            .apply(signal_value)
        )

    else:

        # IMPORTANT:
        # No source evidence = Not Available
        signal_df[standard_name] = "Not Available"


# ------------------------------------------------------------
# 4. SPLIT HIRING / GROWTH SIGNALS
# ------------------------------------------------------------

def split_combined_signal(value, keyword):

    if pd.isna(value):
        return "Not Available"

    value = str(value).strip()

    if value.lower() == "not available":
        return "Not Available"

    lower_value = value.lower()

    if keyword.lower() in lower_value:
        return value

    return "Not Available"


signal_df["Hiring_Signals"] = (
    signal_df["Hiring_Growth_Signals"]
    .apply(
        lambda x:
        split_combined_signal(x, "hiring")
    )
)


signal_df["Growth_Signals"] = (
    signal_df["Hiring_Growth_Signals"]
    .apply(
        lambda x:
        split_combined_signal(x, "growth")
    )
)


signal_df["Expansion_Signals"] = (
    signal_df["Funding_Expansion_Signals"]
    .apply(
        lambda x:
        split_combined_signal(x, "expansion")
    )
)


signal_df["Funding_Signals"] = (
    signal_df["Funding_Expansion_Signals"]
    .apply(
        lambda x:
        split_combined_signal(x, "fund")
    )
)


# ------------------------------------------------------------
# 5. SIGNAL AVAILABILITY FLAGS
# ------------------------------------------------------------

technology_signal_fields = [
    "AI_ML_Signals",
    "Cloud_Data_Signals",
    "API_Platform_Signals",
    "Automation_Workflow_Signals",
    "Software_Development_Signals"
]

business_signal_fields = [
    "Hiring_Signals",
    "Growth_Signals",
    "Expansion_Signals",
    "Funding_Signals",
    "Product_Launch_Signals",
    "Partnership_Signals"
]


all_signal_fields = (
    technology_signal_fields
    + business_signal_fields
)


for col in all_signal_fields:

    signal_df[
        col + "_Available"
    ] = (
        signal_df[col]
        .fillna("Not Available")
        .astype(str)
        .str.lower()
        .ne("not available")
    )


# ------------------------------------------------------------
# 6. OBSERVED SIGNAL COUNT
# ------------------------------------------------------------

availability_columns = [
    col + "_Available"
    for col in all_signal_fields
]

signal_df["Observed_Signal_Count_Workflow"] = (
    signal_df[availability_columns]
    .sum(axis=1)
)


# ------------------------------------------------------------
# 7. TECHNOLOGY SIGNAL COUNT
# ------------------------------------------------------------

technology_availability = [
    col + "_Available"
    for col in technology_signal_fields
]

signal_df["Technology_Signal_Count_Workflow"] = (
    signal_df[technology_availability]
    .sum(axis=1)
)


# ------------------------------------------------------------
# 8. BUSINESS SIGNAL COUNT
# ------------------------------------------------------------

business_availability = [
    col + "_Available"
    for col in business_signal_fields
]

signal_df["Business_Signal_Count_Workflow"] = (
    signal_df[business_availability]
    .sum(axis=1)
)


# ------------------------------------------------------------
# 9. OBSERVED INFORMATION
# ------------------------------------------------------------

def build_observed_information(row):

    observations = []

    for field in all_signal_fields:

        value = row.get(field)

        if (
            pd.notna(value)
            and str(value).strip().lower()
            != "not available"
        ):

            observations.append(
                f"{field.replace('_', ' ')}: {value}"
            )

    if not observations:
        return "No observed signal evidence available"

    return " | ".join(observations)


signal_df["Observed_Information_Workflow"] = (
    signal_df.apply(
        build_observed_information,
        axis=1
    )
)


# ------------------------------------------------------------
# 10. SIGNAL STATUS
# ------------------------------------------------------------

signal_df["Signal_Status_Workflow"] = np.where(
    signal_df["Observed_Signal_Count_Workflow"] > 0,
    "Observed Evidence Available",
    "No Observed Evidence"
)


# ------------------------------------------------------------
# 11. SIGNAL SUMMARY
# ------------------------------------------------------------

signal_summary_rows = []

for field in all_signal_fields:

    available_count = (
        signal_df[field]
        .astype(str)
        .str.lower()
        .ne("not available")
        .sum()
    )

    signal_summary_rows.append({
        "Signal": field,
        "Companies_With_Evidence": available_count,
        "Companies_Without_Evidence":
            len(signal_df) - available_count,
        "Evidence_Percentage":
            round(
                available_count
                / len(signal_df)
                * 100,
                2
            )
    })


signal_summary = pd.DataFrame(
    signal_summary_rows
)

print("\nSignal Summary:")
display(signal_summary)


# ------------------------------------------------------------
# 12. TECHNOLOGY SUMMARY
# ------------------------------------------------------------

technology_summary = pd.DataFrame({
    "Technology_Signal": technology_signal_fields,
    "Companies_With_Evidence": [
        signal_df[col + "_Available"].sum()
        for col in technology_signal_fields
    ]
})

technology_summary["Percentage"] = (
    technology_summary["Companies_With_Evidence"]
    / len(signal_df)
    * 100
).round(2)

print("\nTechnology Signal Summary:")
display(technology_summary)


# ------------------------------------------------------------
# 13. BUSINESS SIGNAL SUMMARY
# ------------------------------------------------------------

business_summary = pd.DataFrame({
    "Business_Signal": business_signal_fields,
    "Companies_With_Evidence": [
        signal_df[col + "_Available"].sum()
        for col in business_signal_fields
    ]
})

business_summary["Percentage"] = (
    business_summary["Companies_With_Evidence"]
    / len(signal_df)
    * 100
).round(2)

print("\nBusiness Signal Summary:")
display(business_summary)


# ------------------------------------------------------------
# 14. SAVE SIGNAL REPORTS
# ------------------------------------------------------------

signal_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Signal_Availability.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

technology_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Technology_Signal_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

business_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Business_Signal_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 15. SAVE STEP OUTPUT
# ------------------------------------------------------------

step8_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_08_Signals.csv"
)

signal_df.to_csv(
    step8_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 8 COMPLETE")
print("=" * 65)

print("Rows preserved:", len(signal_df))
print(
    "Average observed signals:",
    round(
        signal_df[
            "Observed_Signal_Count_Workflow"
        ].mean(),
        2
    )
)

print("\nTechnology evidence:")
for _, row in technology_summary.iterrows():
    print(
        f"{row['Technology_Signal']}: "
        f"{row['Companies_With_Evidence']}"
    )

print("\nBusiness evidence:")
for _, row in business_summary.iterrows():
    print(
        f"{row['Business_Signal']}: "
        f"{row['Companies_With_Evidence']}"
    )

print("\nOutput:")
print(step8_output)
```

```python
# ============================================================
# STEP 10 — PERSONA CLASSIFICATION & VERIFICATION
# ============================================================

persona_df = technology_df.copy()

print("Rows entering Step 10:", len(persona_df))


# ------------------------------------------------------------
# 1. FIND PERSON / TITLE COLUMNS
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


contact_col = find_column(
    persona_df,
    [
        "Contact",
        "Contact_Name",
        "ContactName",
        "Person_Name",
        "PersonName"
    ]
)

title_col = find_column(
    persona_df,
    [
        "Title",
        "Job_Title",
        "JobTitle",
        "Contact_Title",
        "ContactTitle",
        "Designation"
    ]
)

persona_existing_col = find_column(
    persona_df,
    [
        "Persona",
        "Persona_Category",
        "Contact_Persona"
    ]
)

persona_status_existing_col = find_column(
    persona_df,
    [
        "Persona_Data_Status",
        "Persona_Status"
    ]
)


print("\nDetected persona fields:")
print("Contact :", contact_col)
print("Title   :", title_col)
print("Persona :", persona_existing_col)
print("Status  :", persona_status_existing_col)


# ------------------------------------------------------------
# 2. CREATE PERSONA EVIDENCE TEXT
# ------------------------------------------------------------

persona_text_parts = []

for col in [
    contact_col,
    title_col,
    persona_existing_col
]:

    if col and col not in persona_text_parts:
        persona_text_parts.append(col)


if persona_text_parts:

    persona_df[
        "Persona_Evidence_Text_Workflow"
    ] = (
        persona_df[
            persona_text_parts
        ]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
        .str.strip()
    )

else:

    persona_df[
        "Persona_Evidence_Text_Workflow"
    ] = ""


# ------------------------------------------------------------
# 3. PERSONA CLASSIFICATION RULES
# ------------------------------------------------------------

persona_groups = {

    "Decision Maker": [
        "founder",
        "co-founder",
        "cofounder",
        "ceo",
        "chief executive officer",
        "cto",
        "chief technology officer",
        "cio",
        "chief information officer",
        "coo",
        "chief operating officer",
        "cpo",
        "chief product officer"
    ],

    "Technology": [
        "vp engineering",
        "vp of engineering",
        "vice president engineering",
        "vice president of engineering",
        "head engineering",
        "head of engineering",
        "engineering manager",
        "head technology",
        "head of technology",
        "solution architect",
        "technical architect",
        "technology director",
        "engineering director"
    ],

    "Business/Product": [
        "product manager",
        "senior product manager",
        "product director",
        "head of product",
        "vp product",
        "business development",
        "business director",
        "operations director",
        "head of operations",
        "strategy"
    ]
}


# ------------------------------------------------------------
# 4. CLASSIFICATION FUNCTION
# ------------------------------------------------------------

def classify_persona(value):

    if pd.isna(value):
        return "Needs Verification"

    text = str(value).lower().strip()

    if not text:
        return "Needs Verification"

    # Decision Maker
    for term in persona_groups["Decision Maker"]:

        if re.search(
            r"(?<![a-z])"
            + re.escape(term)
            + r"(?![a-z])",
            text
        ):
            return "Decision Maker"


    # Technology
    for term in persona_groups["Technology"]:

        if re.search(
            r"(?<![a-z])"
            + re.escape(term)
            + r"(?![a-z])",
            text
        ):
            return "Technology"


    # Business/Product
    for term in persona_groups["Business/Product"]:

        if re.search(
            r"(?<![a-z])"
            + re.escape(term)
            + r"(?![a-z])",
            text
        ):
            return "Business/Product"


    return "Other"


# ------------------------------------------------------------
# 5. APPLY PERSONA CLASSIFICATION
# ------------------------------------------------------------

persona_df["Persona_Workflow"] = (
    persona_df[
        "Persona_Evidence_Text_Workflow"
    ].apply(classify_persona)
)


# ------------------------------------------------------------
# 6. PERSONA DATA STATUS
# ------------------------------------------------------------

persona_df[
    "Persona_Data_Status_Workflow"
] = np.where(
    persona_df["Persona_Workflow"]
    == "Needs Verification",
    "Needs Verification",
    "Observed"
)


# ------------------------------------------------------------
# 7. PERSONA EVIDENCE STATUS
# ------------------------------------------------------------

persona_df[
    "Persona_Evidence_Status_Workflow"
] = np.where(
    persona_df["Persona_Workflow"]
    == "Needs Verification",
    "No Reliable Persona Evidence",
    "Observed Persona Evidence"
)


# ------------------------------------------------------------
# 8. PERSONA SUMMARY
# ------------------------------------------------------------

persona_summary = (
    persona_df[
        "Persona_Workflow"
    ]
    .value_counts()
    .rename_axis("Persona")
    .reset_index(name="Count")
)

persona_summary["Percentage"] = (
    persona_summary["Count"]
    / len(persona_df)
    * 100
).round(2)


print("\nPersona Summary:")
display(persona_summary)


# ------------------------------------------------------------
# 9. PERSONA STATUS SUMMARY
# ------------------------------------------------------------

persona_status_summary = (
    persona_df[
        "Persona_Data_Status_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Persona_Data_Status"
    )
    .reset_index(name="Count")
)

persona_status_summary["Percentage"] = (
    persona_status_summary["Count"]
    / len(persona_df)
    * 100
).round(2)


print("\nPersona Data Status:")
display(persona_status_summary)


# ------------------------------------------------------------
# 10. PERSONA EVIDENCE CHECK
# ------------------------------------------------------------

print("\nPersona evidence coverage:")

observed_personas = (
    persona_df[
        "Persona_Data_Status_Workflow"
    ]
    .eq("Observed")
    .sum()
)

verification_personas = (
    persona_df[
        "Persona_Data_Status_Workflow"
    ]
    .eq("Needs Verification")
    .sum()
)

print(
    "Observed persona records       :",
    observed_personas
)

print(
    "Needs verification             :",
    verification_personas
)

print(
    "Total records                  :",
    len(persona_df)
)


# ------------------------------------------------------------
# 11. DO NOT FABRICATE PERSONA
# ------------------------------------------------------------

# Explicit audit field
persona_df[
    "Persona_Fabrication_Check"
] = np.where(
    persona_df[
        "Persona_Data_Status_Workflow"
    ] == "Observed",
    "Evidence Based",
    "No Persona Assigned"
)


# ------------------------------------------------------------
# 12. SAVE REPORTS
# ------------------------------------------------------------

persona_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Persona_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

persona_status_summary.to_csv(
    os.path.join(
        QUALITY_DIR,
        "Reusable_Workflow_Persona_Data_Status.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 13. SAVE STEP OUTPUT
# ------------------------------------------------------------

step10_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_10_Persona.csv"
)

persona_df.to_csv(
    step10_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 10 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(persona_df)
)

print(
    "Observed persona:",
    observed_personas
)

print(
    "Needs verification:",
    verification_personas
)

print("\nOutput:")
print(step10_output)
```

```python
# ============================================================
# STEP 11 — NEXZIBYTE SERVICE MAPPING
# ============================================================

service_df = persona_df.copy()

print("Rows entering Step 11:", len(service_df))


# ------------------------------------------------------------
# 1. SERVICE MAPPING RULES
# ------------------------------------------------------------

SERVICE_RULES = {

    "AI / AI Agents": [
        "AI_ML_Signals",
        "AI_Technology_Evidence"
    ],

    "Cloud / Data Engineering": [
        "Cloud_Data_Signals",
        "Cloud_Data_Technology_Evidence"
    ],

    "Backend / API Development": [
        "API_Platform_Signals",
        "API_Platform_Technology_Evidence"
    ],

    "Custom Software Development": [
        "Software_Development_Signals",
        "Software_Development_Technology_Evidence"
    ],

    "Automation / Workflow": [
        "Automation_Workflow_Signals",
        "Automation_Workflow_Technology_Evidence"
    ],

    "Web Development": [
        "Web_Technology_Evidence"
    ],

    "Mobile Development": [
        "Mobile_Technology_Evidence"
    ],

    "Data Analytics": [
        "Data_Analytics_Technology_Evidence"
    ],

    "Dedicated Development Team": [
        "Software_Development_Signals",
        "Hiring_Signals",
        "Growth_Signals",
        "Expansion_Signals"
    ]
}


# ------------------------------------------------------------
# 2. CHECK AVAILABLE EVIDENCE COLUMNS
# ------------------------------------------------------------

available_service_evidence = {}

for service, evidence_fields in SERVICE_RULES.items():

    available_fields = [
        field
        for field in evidence_fields
        if field in service_df.columns
    ]

    available_service_evidence[service] = available_fields


print("\nAvailable service evidence:")

for service, fields in available_service_evidence.items():
    print(f"{service:<35} → {fields}")


# ------------------------------------------------------------
# 3. EVIDENCE DETECTION
# ------------------------------------------------------------

def has_observed_evidence(row, fields):

    for field in fields:

        if field not in row.index:
            continue

        value = row[field]

        if pd.isna(value):
            continue

        value = str(value).strip().lower()

        if value in [
            "",
            "not available",
            "nan",
            "none",
            "null"
        ]:
            continue

        return True

    return False


# ------------------------------------------------------------
# 4. CALCULATE SERVICE EVIDENCE SCORES
# ------------------------------------------------------------

service_score_columns = []

for service, fields in SERVICE_RULES.items():

    safe_name = re.sub(
        r"[^A-Za-z0-9]+",
        "_",
        service
    ).strip("_")

    score_col = (
        "Service_Evidence_"
        + safe_name
        + "_Score"
    )

    service_df[score_col] = service_df.apply(
        lambda row:
        sum(
            has_observed_evidence(
                row,
                [field]
            )
            for field in fields
        ),
        axis=1
    )

    service_score_columns.append(score_col)


# ------------------------------------------------------------
# 5. PRIMARY SERVICE
# ------------------------------------------------------------

def determine_primary_service(row):

    scores = {}

    for service, fields in SERVICE_RULES.items():

        safe_name = re.sub(
            r"[^A-Za-z0-9]+",
            "_",
            service
        ).strip("_")

        score_col = (
            "Service_Evidence_"
            + safe_name
            + "_Score"
        )

        scores[service] = row[score_col]

    max_score = max(scores.values())

    if max_score == 0:
        return "No Evidence Available"

    # Preserve the first service when there is a tie.
    # This makes the workflow deterministic.
    for service, score in scores.items():

        if score == max_score:
            return service

    return "No Evidence Available"


service_df["Primary_Service_Opportunity_Workflow"] = (
    service_df.apply(
        determine_primary_service,
        axis=1
    )
)


# ------------------------------------------------------------
# 6. SECONDARY SERVICES
# ------------------------------------------------------------

def determine_secondary_services(row):

    scores = {}

    for service, fields in SERVICE_RULES.items():

        safe_name = re.sub(
            r"[^A-Za-z0-9]+",
            "_",
            service
        ).strip("_")

        score_col = (
            "Service_Evidence_"
            + safe_name
            + "_Score"
        )

        scores[service] = row[score_col]

    ranked = sorted(
        scores.items(),
        key=lambda x: x[1],
        reverse=True
    )

    primary = row[
        "Primary_Service_Opportunity_Workflow"
    ]

    secondary = []

    for service, score in ranked:

        if service == primary:
            continue

        if score > 0:
            secondary.append(service)

        if len(secondary) == 2:
            break

    if not secondary:
        return "Not Available"

    return "; ".join(secondary)


service_df[
    "Secondary_Service_Opportunity_Workflow"
] = service_df.apply(
    determine_secondary_services,
    axis=1
)


# ------------------------------------------------------------
# 7. SERVICE REASON
# ------------------------------------------------------------

def service_reason(row):

    primary = row[
        "Primary_Service_Opportunity_Workflow"
    ]

    if primary == "No Evidence Available":
        return (
            "No observed technology/business evidence "
            "available for service mapping"
        )

    service_fields = SERVICE_RULES.get(
        primary,
        []
    )

    evidence = []

    for field in service_fields:

        if field not in row.index:
            continue

        value = row[field]

        if pd.isna(value):
            continue

        value = str(value).strip()

        if value.lower() == "not available":
            continue

        if value:
            evidence.append(
                f"{field.replace('_', ' ')}: {value}"
            )

    if evidence:

        return (
            "Potential opportunity inferred from "
            "observed evidence: "
            + " | ".join(evidence)
        )

    return (
        "Potential opportunity inferred from "
        "available business/technology evidence"
    )


service_df[
    "Service_Opportunity_Reason_Workflow"
] = service_df.apply(
    service_reason,
    axis=1
)


# ------------------------------------------------------------
# 8. OBSERVED VS INFERRED LABEL
# ------------------------------------------------------------

service_df[
    "Service_Opportunity_Type_Workflow"
] = np.where(
    service_df[
        "Primary_Service_Opportunity_Workflow"
    ]
    == "No Evidence Available",
    "No Evidence",
    "Inferred Opportunity"
)


# ------------------------------------------------------------
# 9. SERVICE SUMMARY
# ------------------------------------------------------------

primary_service_summary = (
    service_df[
        "Primary_Service_Opportunity_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Primary_Service"
    )
    .reset_index(
        name="Company_Count"
    )
)

primary_service_summary["Percentage"] = (
    primary_service_summary["Company_Count"]
    / len(service_df)
    * 100
).round(2)


print("\nPrimary Service Opportunity:")
display(primary_service_summary)


# ------------------------------------------------------------
# 10. SECONDARY SERVICE SUMMARY
# ------------------------------------------------------------

secondary_service_summary = (
    service_df[
        "Secondary_Service_Opportunity_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Secondary_Service"
    )
    .reset_index(
        name="Company_Count"
    )
)

print("\nSecondary Service Opportunity:")
display(
    secondary_service_summary.head(20)
)


# ------------------------------------------------------------
# 11. SERVICE BY COMPANY SIZE
# ------------------------------------------------------------

if "Company_Size_Category_Workflow" in service_df.columns:

    service_company_size = pd.crosstab(
        service_df[
            "Company_Size_Category_Workflow"
        ],
        service_df[
            "Primary_Service_Opportunity_Workflow"
        ]
    )

    print("\nService by Company Size:")
    display(service_company_size)


# ------------------------------------------------------------
# 12. SERVICE BY INDUSTRY
# ------------------------------------------------------------

if "Industry_Standardized_Workflow" in service_df.columns:

    service_industry = pd.crosstab(
        service_df[
            "Industry_Standardized_Workflow"
        ],
        service_df[
            "Primary_Service_Opportunity_Workflow"
        ]
    )

    print("\nService by Industry:")
    display(
        service_industry.head(20)
    )


# ------------------------------------------------------------
# 13. SAVE REPORTS
# ------------------------------------------------------------

primary_service_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Primary_Service_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

secondary_service_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Secondary_Service_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


if "Company_Size_Category_Workflow" in service_df.columns:

    service_company_size.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Service_By_Company_Size.csv"
        ),
        encoding="utf-8-sig"
    )


if "Industry_Standardized_Workflow" in service_df.columns:

    service_industry.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Service_By_Industry.csv"
        ),
        encoding="utf-8-sig"
    )


# ------------------------------------------------------------
# 14. SAVE STEP OUTPUT
# ------------------------------------------------------------

step11_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_11_Service_Mapping.csv"
)

service_df.to_csv(
    step11_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 11 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(service_df)
)

print(
    "Companies with mapped service:",
    (
        service_df[
            "Primary_Service_Opportunity_Workflow"
        ]
        != "No Evidence Available"
    ).sum()
)

print(
    "Companies without service evidence:",
    (
        service_df[
            "Primary_Service_Opportunity_Workflow"
        ]
        == "No Evidence Available"
    ).sum()
)

print("\nOutput:")
print(step11_output)
```

```python
# ============================================================
# STEP 12 — MARKETING SEGMENTATION
# ============================================================

segment_df = service_df.copy()

print("Rows entering Step 12:", len(segment_df))


# ------------------------------------------------------------
# 1. HELPER FUNCTIONS
# ------------------------------------------------------------

def get_value(row, column, default="Not Available"):

    if column not in row.index:
        return default

    value = row[column]

    if pd.isna(value):
        return default

    value = str(value).strip()

    if not value:
        return default

    return value


def evidence_available(row, column):

    if column not in row.index:
        return False

    value = row[column]

    if pd.isna(value):
        return False

    return str(value).strip().lower() not in [
        "",
        "not available",
        "nan",
        "none",
        "null"
    ]


# ------------------------------------------------------------
# 2. COMPANY PROFILE
# ------------------------------------------------------------

industry_col = (
    "Industry_Standardized_Workflow"
    if "Industry_Standardized_Workflow" in segment_df.columns
    else None
)

company_size_col = (
    "Company_Size_Category_Workflow"
    if "Company_Size_Category_Workflow" in segment_df.columns
    else None
)

business_model_col = (
    "Business_Model_Workflow"
    if "Business_Model_Workflow" in segment_df.columns
    else None
)

persona_col = (
    "Persona_Workflow"
    if "Persona_Workflow" in segment_df.columns
    else None
)


# ------------------------------------------------------------
# 3. CREATE SIGNAL FLAGS
# ------------------------------------------------------------

signal_flag_map = {
    "AI": "AI_ML_Signals_Available",
    "Cloud/Data": "Cloud_Data_Signals_Available",
    "API/Platform": "API_Platform_Signals_Available",
    "Automation": "Automation_Workflow_Signals_Available",
    "Software": "Software_Development_Signals_Available",
    "Hiring": "Hiring_Signals_Available",
    "Growth": "Growth_Signals_Available",
    "Expansion": "Expansion_Signals_Available",
    "Funding": "Funding_Signals_Available",
    "Product Launch": "Product_Launch_Signals_Available",
    "Partnership": "Partnership_Signals_Available"
}


# ------------------------------------------------------------
# 4. CREATE MARKETING SEGMENT
# ------------------------------------------------------------

def determine_marketing_segment(row):

    industry = get_value(
        row,
        industry_col,
        "Not Available"
    ) if industry_col else "Not Available"

    company_size = get_value(
        row,
        company_size_col,
        "Not Available"
    ) if company_size_col else "Not Available"

    business_model = get_value(
        row,
        business_model_col,
        "Not Available"
    ) if business_model_col else "Not Available"

    persona = get_value(
        row,
        persona_col,
        "Needs Verification"
    ) if persona_col else "Needs Verification"


    # --------------------------------------------------------
    # OBSERVED SIGNAL FLAGS
    # --------------------------------------------------------

    ai = evidence_available(
        row,
        signal_flag_map["AI"]
    )

    cloud = evidence_available(
        row,
        signal_flag_map["Cloud/Data"]
    )

    api = evidence_available(
        row,
        signal_flag_map["API/Platform"]
    )

    automation = evidence_available(
        row,
        signal_flag_map["Automation"]
    )

    software = evidence_available(
        row,
        signal_flag_map["Software"]
    )

    hiring = evidence_available(
        row,
        signal_flag_map["Hiring"]
    )

    growth = evidence_available(
        row,
        signal_flag_map["Growth"]
    )

    expansion = evidence_available(
        row,
        signal_flag_map["Expansion"]
    )

    funding = evidence_available(
        row,
        signal_flag_map["Funding"]
    )

    product_launch = evidence_available(
        row,
        signal_flag_map["Product Launch"]
    )

    partnership = evidence_available(
        row,
        signal_flag_map["Partnership"]
    )


    # --------------------------------------------------------
    # PERSONA + SIGNAL SEGMENTS
    # --------------------------------------------------------

    if (
        industry.lower() == "saas"
        and persona == "Technology"
        and ai
    ):
        return "SaaS + Technology + AI"

    if (
        industry.lower() == "saas"
        and persona == "Decision Maker"
        and software
    ):
        return "SaaS + Decision Maker + Software"


    # --------------------------------------------------------
    # AI / ML
    # --------------------------------------------------------

    if ai and cloud and api:
        return "AI/ML + Cloud/Data + API/Platform"

    if ai:
        return "AI/ML"


    # --------------------------------------------------------
    # DATA / CLOUD
    # --------------------------------------------------------

    if cloud and api and software:
        return "Cloud/Data + API/Platform + Software"

    if cloud:
        return "Cloud/Data"


    # --------------------------------------------------------
    # API / PLATFORM
    # --------------------------------------------------------

    if api:
        return "API/Platform"


    # --------------------------------------------------------
    # AUTOMATION
    # --------------------------------------------------------

    if automation:
        return "Automation/Workflow"


    # --------------------------------------------------------
    # SOFTWARE
    # --------------------------------------------------------

    if software and hiring:
        return "Software + Hiring"

    if software:
        return "Software Development"


    # --------------------------------------------------------
    # BUSINESS GROWTH SIGNALS
    # --------------------------------------------------------

    if hiring and growth:
        return "Hiring + Growth"

    if funding and expansion:
        return "Funding + Expansion"

    if product_launch:
        return "Product Launch"

    if partnership:
        return "Partnership"


    # --------------------------------------------------------
    # COMPANY SIZE SEGMENTS
    # --------------------------------------------------------

    if (
        company_size == "SME"
        and automation
    ):
        return "SME + Automation"

    if (
        company_size in [
            "Enterprise",
            "Large Enterprise"
        ]
        and software
    ):
        return "Enterprise + Software"


    # --------------------------------------------------------
    # GENERAL
    # --------------------------------------------------------

    if industry != "Not Available":
        return "Industry Prospect"

    return "General Prospect"


segment_df["Marketing_Segment_Workflow"] = (
    segment_df.apply(
        determine_marketing_segment,
        axis=1
    )
)


# ------------------------------------------------------------
# 5. SEGMENT REASON
# ------------------------------------------------------------

def build_segment_reason(row):

    segment = row[
        "Marketing_Segment_Workflow"
    ]

    reasons = []

    if industry_col:
        industry = get_value(
            row,
            industry_col
        )

        if industry != "Not Available":
            reasons.append(
                f"Industry: {industry}"
            )

    if company_size_col:
        size = get_value(
            row,
            company_size_col
        )

        if size != "Not Available":
            reasons.append(
                f"Company Size: {size}"
            )

    if persona_col:
        persona = get_value(
            row,
            persona_col
        )

        if persona != "Needs Verification":
            reasons.append(
                f"Persona: {persona}"
            )

    observed_signals = []

    for signal_name, flag_col in signal_flag_map.items():

        if evidence_available(
            row,
            flag_col
        ):
            observed_signals.append(
                signal_name
            )

    if observed_signals:
        reasons.append(
            "Observed Signals: "
            + ", ".join(observed_signals)
        )

    service = get_value(
        row,
        "Primary_Service_Opportunity_Workflow"
    )

    if service != "No Evidence Available":
        reasons.append(
            f"Potential Service: {service}"
        )

    if not reasons:
        return (
            "No sufficient observed evidence "
            "for detailed segmentation"
        )

    return (
        f"Segment '{segment}' created from "
        + " | ".join(reasons)
    )


segment_df["Marketing_Segment_Reason_Workflow"] = (
    segment_df.apply(
        build_segment_reason,
        axis=1
    )
)


# ------------------------------------------------------------
# 6. SEGMENT EVIDENCE TYPE
# ------------------------------------------------------------

segment_df[
    "Marketing_Segment_Evidence_Type_Workflow"
] = np.where(
    segment_df[
        "Marketing_Segment_Workflow"
    ].isin([
        "Industry Prospect",
        "General Prospect"
    ]),
    "Limited Observed Evidence",
    "Observed Evidence Based"
)


# ------------------------------------------------------------
# 7. SEGMENT PRIORITY
# ------------------------------------------------------------

segment_df[
    "Marketing_Segment_Priority_Workflow"
] = "Standard"


high_priority_segments = [
    "AI/ML",
    "AI/ML + Cloud/Data + API/Platform",
    "SaaS + Technology + AI",
    "SaaS + Decision Maker + Software",
    "Software + Hiring",
    "Hiring + Growth",
    "Funding + Expansion"
]

segment_df.loc[
    segment_df[
        "Marketing_Segment_Workflow"
    ].isin(high_priority_segments),
    "Marketing_Segment_Priority_Workflow"
] = "High Potential"


# ------------------------------------------------------------
# 8. SEGMENT SUMMARY
# ------------------------------------------------------------

segment_summary = (
    segment_df[
        "Marketing_Segment_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Marketing_Segment"
    )
    .reset_index(
        name="Company_Count"
    )
)

segment_summary["Percentage"] = (
    segment_summary["Company_Count"]
    / len(segment_df)
    * 100
).round(2)


print("\nMarketing Segment Summary:")
display(segment_summary)


# ------------------------------------------------------------
# 9. SEGMENT BY PRIORITY
# ------------------------------------------------------------

if "Lead_Priority_V2" in segment_df.columns:

    segment_priority = pd.crosstab(
        segment_df[
            "Marketing_Segment_Workflow"
        ],
        segment_df[
            "Lead_Priority_V2"
        ]
    )

elif "Lead_Priority" in segment_df.columns:

    segment_priority = pd.crosstab(
        segment_df[
            "Marketing_Segment_Workflow"
        ],
        segment_df[
            "Lead_Priority"
        ]
    )

else:

    segment_priority = pd.DataFrame()


if not segment_priority.empty:

    print("\nMarketing Segment by Lead Priority:")
    display(segment_priority)


# ------------------------------------------------------------
# 10. SEGMENT BY SERVICE
# ------------------------------------------------------------

if (
    "Primary_Service_Opportunity_Workflow"
    in segment_df.columns
):

    segment_service = pd.crosstab(
        segment_df[
            "Marketing_Segment_Workflow"
        ],
        segment_df[
            "Primary_Service_Opportunity_Workflow"
        ]
    )

    print("\nMarketing Segment by Primary Service:")
    display(
        segment_service.head(20)
    )

else:

    segment_service = pd.DataFrame()


# ------------------------------------------------------------
# 11. SEGMENT BY INDUSTRY
# ------------------------------------------------------------

if industry_col:

    segment_industry = pd.crosstab(
        segment_df[
            "Marketing_Segment_Workflow"
        ],
        segment_df[
            industry_col
        ]
    )

    print("\nMarketing Segment by Industry:")
    display(
        segment_industry.head(20)
    )

else:

    segment_industry = pd.DataFrame()


# ------------------------------------------------------------
# 12. SAVE REPORTS
# ------------------------------------------------------------

segment_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Marketing_Segment_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


if not segment_priority.empty:

    segment_priority.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Segment_By_Priority.csv"
        ),
        encoding="utf-8-sig"
    )


if not segment_service.empty:

    segment_service.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Segment_By_Service.csv"
        ),
        encoding="utf-8-sig"
    )


if not segment_industry.empty:

    segment_industry.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Segment_By_Industry.csv"
        ),
        encoding="utf-8-sig"
    )


# ------------------------------------------------------------
# 13. SAVE STEP OUTPUT
# ------------------------------------------------------------

step12_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_12_Marketing_Segmentation.csv"
)

segment_df.to_csv(
    step12_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 12 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(segment_df)
)

print(
    "Unique marketing segments:",
    segment_df[
        "Marketing_Segment_Workflow"
    ].nunique()
)

print(
    "Missing marketing segments:",
    segment_df[
        "Marketing_Segment_Workflow"
    ].isna().sum()
)

print("\nOutput:")
print(step12_output)
```

```python
# ============================================================
# STEP 13 — MARKETING READINESS CLASSIFICATION
# ============================================================

readiness_df = segment_df.copy()

print("Rows entering Step 13:", len(readiness_df))


# ------------------------------------------------------------
# 1. HELPER FUNCTIONS
# ------------------------------------------------------------

def find_column(df, possible_names):

    normalized = {
        re.sub(r"[^a-z0-9]", "", str(col).lower()): col
        for col in df.columns
    }

    for name in possible_names:

        key = re.sub(
            r"[^a-z0-9]",
            "",
            name.lower()
        )

        if key in normalized:
            return normalized[key]

    return None


def safe_text(row, column):

    if column not in row.index:
        return ""

    value = row[column]

    if pd.isna(value):
        return ""

    return str(value).strip()


# ------------------------------------------------------------
# 2. IDENTIFY IMPORTANT FIELDS
# ------------------------------------------------------------

company_col = find_column(
    readiness_df,
    [
        "Company_Name",
        "Company",
        "CompanyName"
    ]
)

website_col = find_column(
    readiness_df,
    [
        "Website",
        "Company_Website",
        "CompanyWebsite"
    ]
)

email_col = find_column(
    readiness_df,
    [
        "Email",
        "Email_Address",
        "EmailAddress"
    ]
)

data_quality_col = find_column(
    readiness_df,
    [
        "Data_Quality_Status",
        "Data_Quality_Status_V2"
    ]
)

duplicate_decision_col = find_column(
    readiness_df,
    [
        "Duplicate_Decision"
    ]
)

persona_status_col = (
    "Persona_Data_Status_Workflow"
    if "Persona_Data_Status_Workflow"
    in readiness_df.columns
    else None
)

signal_count_col = (
    "Observed_Signal_Count_Workflow"
    if "Observed_Signal_Count_Workflow"
    in readiness_df.columns
    else None
)

technology_count_col = (
    "Technology_Evidence_Count_Workflow"
    if "Technology_Evidence_Count_Workflow"
    in readiness_df.columns
    else None
)

service_col = (
    "Primary_Service_Opportunity_Workflow"
    if "Primary_Service_Opportunity_Workflow"
    in readiness_df.columns
    else None
)


print("\nDetected readiness fields:")
print("Company:", company_col)
print("Website:", website_col)
print("Email:", email_col)
print("Data Quality:", data_quality_col)
print("Duplicate Decision:", duplicate_decision_col)
print("Persona Status:", persona_status_col)
print("Signal Count:", signal_count_col)
print("Technology Evidence:", technology_count_col)
print("Service:", service_col)


# ------------------------------------------------------------
# 3. CREATE READINESS COMPONENTS
# ------------------------------------------------------------

readiness_df[
    "Readiness_Company_Name_Check"
] = True

readiness_df[
    "Readiness_Website_Check"
] = False

readiness_df[
    "Readiness_Email_Check"
] = False

readiness_df[
    "Readiness_Data_Quality_Check"
] = True

readiness_df[
    "Readiness_Duplicate_Check"
] = True


# ------------------------------------------------------------
# 4. COMPANY NAME CHECK
# ------------------------------------------------------------

if company_col:

    readiness_df[
        "Readiness_Company_Name_Check"
    ] = (
        readiness_df[company_col]
        .notna()
        &
        (
            readiness_df[company_col]
            .astype(str)
            .str.strip()
            .str.len()
            >= 2
        )
    )


# ------------------------------------------------------------
# 5. WEBSITE CHECK
# ------------------------------------------------------------

if website_col:

    readiness_df[
        "Readiness_Website_Check"
    ] = (
        readiness_df[website_col]
        .notna()
        &
        readiness_df[website_col]
        .astype(str)
        .str.contains(
            r"\.",
            regex=True,
            na=False
        )
    )


# ------------------------------------------------------------
# 6. EMAIL CHECK
# ------------------------------------------------------------

if email_col:

    email_pattern = (
        r"^[A-Za-z0-9._%+-]+@"
        r"[A-Za-z0-9.-]+\."
        r"[A-Za-z]{2,}$"
    )

    readiness_df[
        "Readiness_Email_Check"
    ] = (
        readiness_df[email_col]
        .fillna("")
        .astype(str)
        .str.match(
            email_pattern,
            na=False
        )
    )


# ------------------------------------------------------------
# 7. DATA QUALITY CHECK
# ------------------------------------------------------------

if data_quality_col:

    readiness_df[
        "Readiness_Data_Quality_Check"
    ] = ~readiness_df[
        data_quality_col
    ].astype(str).str.lower().isin([
        "invalid / do not use",
        "invalid",
        "duplicate"
    ])


# ------------------------------------------------------------
# 8. DUPLICATE CHECK
# ------------------------------------------------------------

if duplicate_decision_col:

    readiness_df[
        "Readiness_Duplicate_Check"
    ] = (
        readiness_df[
            duplicate_decision_col
        ]
        .astype(str)
        .str.upper()
        .eq("KEEP")
    )


# ------------------------------------------------------------
# 9. EVIDENCE CHECK
# ------------------------------------------------------------

if signal_count_col:

    readiness_df[
        "Readiness_Signal_Evidence"
    ] = (
        pd.to_numeric(
            readiness_df[
                signal_count_col
            ],
            errors="coerce"
        )
        .fillna(0)
        > 0
    )

else:

    readiness_df[
        "Readiness_Signal_Evidence"
    ] = False


# ------------------------------------------------------------
# 10. TECHNOLOGY EVIDENCE CHECK
# ------------------------------------------------------------

if technology_count_col:

    readiness_df[
        "Readiness_Technology_Evidence"
    ] = (
        pd.to_numeric(
            readiness_df[
                technology_count_col
            ],
            errors="coerce"
        )
        .fillna(0)
        > 0
    )

else:

    readiness_df[
        "Readiness_Technology_Evidence"
    ] = False


# ------------------------------------------------------------
# 11. SERVICE OPPORTUNITY CHECK
# ------------------------------------------------------------

if service_col:

    readiness_df[
        "Readiness_Service_Opportunity"
    ] = (
        readiness_df[
            service_col
        ]
        .astype(str)
        .ne("No Evidence Available")
        &
        readiness_df[
            service_col
        ]
        .astype(str)
        .ne("Not Available")
    )

else:

    readiness_df[
        "Readiness_Service_Opportunity"
    ] = False


# ------------------------------------------------------------
# 12. FINAL READINESS CLASSIFICATION
# ------------------------------------------------------------

def determine_readiness(row):

    company_valid = row[
        "Readiness_Company_Name_Check"
    ]

    data_quality_valid = row[
        "Readiness_Data_Quality_Check"
    ]

    duplicate_valid = row[
        "Readiness_Duplicate_Check"
    ]

    website_valid = row[
        "Readiness_Website_Check"
    ]

    email_valid = row[
        "Readiness_Email_Check"
    ]

    signal_evidence = row[
        "Readiness_Signal_Evidence"
    ]

    technology_evidence = row[
        "Readiness_Technology_Evidence"
    ]

    service_opportunity = row[
        "Readiness_Service_Opportunity"
    ]


    # --------------------------------------------------------
    # DO NOT USE
    # --------------------------------------------------------

    if not company_valid:
        return "DO NOT USE"

    if not data_quality_valid:
        return "DO NOT USE"

    if not duplicate_valid:
        return "DO NOT USE"


    # --------------------------------------------------------
    # READY
    # --------------------------------------------------------
    # A usable company identity plus contactability
    # and meaningful evidence.

    if (
        website_valid
        and (
            email_valid
            or signal_evidence
            or technology_evidence
        )
        and (
            signal_evidence
            or technology_evidence
            or service_opportunity
        )
    ):
        return "READY"


    # --------------------------------------------------------
    # EVERYTHING ELSE
    # --------------------------------------------------------

    return "NEEDS VERIFICATION"


readiness_df[
    "Marketing_Readiness_Workflow"
] = readiness_df.apply(
    determine_readiness,
    axis=1
)


# ------------------------------------------------------------
# 13. READINESS REASON
# ------------------------------------------------------------

def readiness_reason(row):

    status = row[
        "Marketing_Readiness_Workflow"
    ]

    reasons = []

    if not row[
        "Readiness_Company_Name_Check"
    ]:
        reasons.append(
            "Invalid/Missing Company Name"
        )

    if not row[
        "Readiness_Website_Check"
    ]:
        reasons.append(
            "Website Missing/Invalid"
        )

    if not row[
        "Readiness_Email_Check"
    ]:
        reasons.append(
            "Email Missing/Invalid"
        )

    if not row[
        "Readiness_Signal_Evidence"
    ]:
        reasons.append(
            "No Observed Business Signal"
        )

    if not row[
        "Readiness_Technology_Evidence"
    ]:
        reasons.append(
            "No Observed Technology Evidence"
        )

    if not row[
        "Readiness_Service_Opportunity"
    ]:
        reasons.append(
            "No Service Evidence"
        )

    if not row[
        "Readiness_Data_Quality_Check"
    ]:
        reasons.append(
            "Data Quality Issue"
        )

    if not row[
        "Readiness_Duplicate_Check"
    ]:
        reasons.append(
            "Duplicate Requires Review"
        )


    if status == "READY":

        return (
            "Company is usable for marketing based on "
            "valid company identity, contactability and "
            "available evidence."
        )

    if status == "DO NOT USE":

        return (
            "Record should not be used until the underlying "
            "data-quality issue is resolved."
        )

    if not reasons:

        return "Additional verification required"

    return "; ".join(reasons)


readiness_df[
    "Marketing_Readiness_Reason_Workflow"
] = readiness_df.apply(
    readiness_reason,
    axis=1
)


# ------------------------------------------------------------
# 14. READINESS SCORE
# ------------------------------------------------------------

readiness_components = [
    "Readiness_Company_Name_Check",
    "Readiness_Website_Check",
    "Readiness_Email_Check",
    "Readiness_Data_Quality_Check",
    "Readiness_Duplicate_Check",
    "Readiness_Signal_Evidence",
    "Readiness_Technology_Evidence",
    "Readiness_Service_Opportunity"
]

readiness_df[
    "Marketing_Readiness_Component_Score"
] = (
    readiness_df[
        readiness_components
    ].sum(axis=1)
)


# ------------------------------------------------------------
# 15. SUMMARY
# ------------------------------------------------------------

readiness_summary = (
    readiness_df[
        "Marketing_Readiness_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Marketing_Readiness"
    )
    .reset_index(
        name="Company_Count"
    )
)

readiness_summary["Percentage"] = (
    readiness_summary["Company_Count"]
    / len(readiness_df)
    * 100
).round(2)


print("\nMarketing Readiness Summary:")
display(readiness_summary)


# ------------------------------------------------------------
# 16. READINESS BY SEGMENT
# ------------------------------------------------------------

if "Marketing_Segment_Workflow" in readiness_df.columns:

    readiness_segment = pd.crosstab(
        readiness_df[
            "Marketing_Segment_Workflow"
        ],
        readiness_df[
            "Marketing_Readiness_Workflow"
        ]
    )

    print("\nMarketing Readiness by Segment:")
    display(readiness_segment)


# ------------------------------------------------------------
# 17. READINESS BY PRIORITY IF AVAILABLE
# ------------------------------------------------------------

priority_col = find_column(
    readiness_df,
    [
        "Lead_Priority_V2",
        "Lead_Priority"
    ]
)

if priority_col:

    readiness_priority = pd.crosstab(
        readiness_df[
            priority_col
        ],
        readiness_df[
            "Marketing_Readiness_Workflow"
        ]
    )

    print("\nMarketing Readiness by Priority:")
    display(readiness_priority)

else:

    readiness_priority = pd.DataFrame()


# ------------------------------------------------------------
# 18. SAVE REPORTS
# ------------------------------------------------------------

readiness_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Marketing_Readiness_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

if "Marketing_Segment_Workflow" in readiness_df.columns:

    readiness_segment.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Readiness_By_Segment.csv"
        ),
        encoding="utf-8-sig"
    )

if not readiness_priority.empty:

    readiness_priority.to_csv(
        os.path.join(
            ANALYSIS_DIR,
            "Reusable_Workflow_Readiness_By_Priority.csv"
        ),
        encoding="utf-8-sig"
    )


# ------------------------------------------------------------
# 19. SAVE STEP OUTPUT
# ------------------------------------------------------------

step13_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_13_Marketing_Readiness.csv"
)

readiness_df.to_csv(
    step13_output,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# FINAL STATUS
# ------------------------------------------------------------

print("=" * 65)
print("STEP 13 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(readiness_df)
)

for _, row in readiness_summary.iterrows():

    print(
        f"{row['Marketing_Readiness']}: "
        f"{row['Company_Count']}"
    )

print("\nOutput:")
print(step13_output)
```

```python
# ============================================================
# STEP 14 — PROSPECT SCORING + LEAD PRIORITY
# ============================================================

scoring_df = readiness_df.copy()

print("Rows entering Step 14:", len(scoring_df))


# ------------------------------------------------------------
# 1. NUMERIC HELPER
# ------------------------------------------------------------

def numeric_value(df, column, default=0):

    if column and column in df.columns:
        return pd.to_numeric(
            df[column],
            errors="coerce"
        ).fillna(default)

    return pd.Series(
        default,
        index=df.index
    )


# ------------------------------------------------------------
# 2. COMPANY FIT SCORE — MAX 20
# ------------------------------------------------------------

company_fit = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

# Company size
size_col = find_column(
    scoring_df,
    [
        "Company_Size_Category_Workflow",
        "Company_Size_Category"
    ]
)

if size_col:

    size_points = {
        "Enterprise": 10,
        "Large Enterprise": 9,
        "Mid-Market": 7,
        "SME": 4,
        "Startup/Small": 2,
        "Needs Verification": 1
    }

    company_fit += (
        scoring_df[size_col]
        .astype(str)
        .map(size_points)
        .fillna(0)
    )


# Industry available
industry_col = find_column(
    scoring_df,
    [
        "Industry_Standardized_Workflow",
        "Industry_Standardized",
        "Industry"
    ]
)

if industry_col:

    company_fit += (
        scoring_df[industry_col]
        .notna()
        .astype(int) * 4
    )


# Location available
country_col = find_column(
    scoring_df,
    [
        "Country_Standardized_Workflow",
        "Country_Standardized",
        "Company_Country"
    ]
)

if country_col:

    company_fit += (
        scoring_df[country_col]
        .notna()
        .astype(int) * 3
    )


# Website available
company_fit += (
    scoring_df[
        "Readiness_Website_Check"
    ]
    .astype(int) * 3
)

company_fit = company_fit.clip(0, 20)


# ------------------------------------------------------------
# 3. TECHNOLOGY FIT SCORE — MAX 25
# ------------------------------------------------------------

technology_fit = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

technology_signal_col = (
    "Technology_Evidence_Count_Workflow"
    if "Technology_Evidence_Count_Workflow"
    in scoring_df.columns
    else None
)

if technology_signal_col:

    technology_fit += (
        pd.to_numeric(
            scoring_df[
                technology_signal_col
            ],
            errors="coerce"
        )
        .fillna(0)
        .clip(0, 5)
        * 5
    )

else:

    technology_fit += (
        scoring_df[
            "Readiness_Technology_Evidence"
        ]
        .astype(int) * 10
    )

technology_fit = technology_fit.clip(0, 25)


# ------------------------------------------------------------
# 4. GROWTH / INTENT SCORE — MAX 20
# ------------------------------------------------------------

growth_intent = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

growth_col = find_column(
    scoring_df,
    [
        "Growth_Signals_Workflow",
        "Hiring_Growth_Signals",
        "Hiring_Signals_Workflow"
    ]
)

if growth_col:

    growth_text = (
        scoring_df[growth_col]
        .fillna("")
        .astype(str)
        .str.lower()
    )

    growth_keywords = [
        "hiring",
        "growth",
        "expansion",
        "expanding",
        "scaling",
        "recruit",
        "job"
    ]

    for keyword in growth_keywords:

        growth_intent += (
            growth_text
            .str.contains(
                keyword,
                regex=False,
                na=False
            )
            .astype(int)
        )

growth_intent = growth_intent.clip(0, 20)


# ------------------------------------------------------------
# 5. FUNDING / EXPANSION SCORE — MAX 15
# ------------------------------------------------------------

funding_expansion = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

funding_col = find_column(
    scoring_df,
    [
        "Funding_Signals_Workflow",
        "Funding_Expansion_Signals"
    ]
)

if funding_col:

    funding_text = (
        scoring_df[funding_col]
        .fillna("")
        .astype(str)
        .str.lower()
    )

    funding_keywords = [
        "funding",
        "funded",
        "investment",
        "raised",
        "series a",
        "series b",
        "series c",
        "venture",
        "expansion"
    ]

    for keyword in funding_keywords:

        funding_expansion += (
            funding_text
            .str.contains(
                keyword,
                regex=False,
                na=False
            )
            .astype(int)
        )

funding_expansion = funding_expansion.clip(0, 15)


# ------------------------------------------------------------
# 6. MARKETING READINESS SCORE — MAX 10
# ------------------------------------------------------------

marketing_readiness = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

marketing_readiness += (
    scoring_df[
        "Readiness_Company_Name_Check"
    ].astype(int) * 2
)

marketing_readiness += (
    scoring_df[
        "Readiness_Website_Check"
    ].astype(int) * 2
)

marketing_readiness += (
    scoring_df[
        "Readiness_Email_Check"
    ].astype(int) * 2
)

marketing_readiness += (
    scoring_df[
        "Readiness_Signal_Evidence"
    ].astype(int) * 2
)

marketing_readiness += (
    scoring_df[
        "Readiness_Technology_Evidence"
    ].astype(int) * 2
)

marketing_readiness = marketing_readiness.clip(0, 10)


# ------------------------------------------------------------
# 7. DATA CONFIDENCE SCORE — MAX 10
# ------------------------------------------------------------

data_confidence = pd.Series(
    0,
    index=scoring_df.index,
    dtype=float
)

data_confidence += (
    scoring_df[
        "Readiness_Company_Name_Check"
    ].astype(int) * 2
)

data_confidence += (
    scoring_df[
        "Readiness_Website_Check"
    ].astype(int) * 2
)

data_confidence += (
    scoring_df[
        "Readiness_Email_Check"
    ].astype(int) * 2
)

data_confidence += (
    scoring_df[
        "Readiness_Data_Quality_Check"
    ].astype(int) * 2
)

data_confidence += (
    scoring_df[
        "Readiness_Technology_Evidence"
    ].astype(int) * 1
)

data_confidence += (
    scoring_df[
        "Readiness_Signal_Evidence"
    ].astype(int) * 1
)

data_confidence = data_confidence.clip(0, 10)


# ------------------------------------------------------------
# 8. STORE COMPONENT SCORES
# ------------------------------------------------------------

scoring_df[
    "Company_Fit_Score_Workflow"
] = company_fit.round(2)

scoring_df[
    "Technology_Fit_Score_Workflow"
] = technology_fit.round(2)

scoring_df[
    "Growth_Intent_Score_Workflow"
] = growth_intent.round(2)

scoring_df[
    "Funding_Expansion_Score_Workflow"
] = funding_expansion.round(2)

scoring_df[
    "Marketing_Readiness_Score_Workflow"
] = marketing_readiness.round(2)

scoring_df[
    "Data_Confidence_Score_Workflow"
] = data_confidence.round(2)


# ------------------------------------------------------------
# 9. FINAL PROSPECT SCORE — MAX 100
# ------------------------------------------------------------

scoring_df[
    "Prospect_Score_Workflow"
] = (
    scoring_df[
        "Company_Fit_Score_Workflow"
    ]
    +
    scoring_df[
        "Technology_Fit_Score_Workflow"
    ]
    +
    scoring_df[
        "Growth_Intent_Score_Workflow"
    ]
    +
    scoring_df[
        "Funding_Expansion_Score_Workflow"
    ]
    +
    scoring_df[
        "Marketing_Readiness_Score_Workflow"
    ]
    +
    scoring_df[
        "Data_Confidence_Score_Workflow"
    ]
).round(2)


# ------------------------------------------------------------
# 10. LEAD PRIORITY
# ------------------------------------------------------------

def assign_priority(row):

    readiness = row[
        "Marketing_Readiness_Workflow"
    ]

    score = row[
        "Prospect_Score_Workflow"
    ]

    # Never prioritize unusable records
    if readiness == "DO NOT USE":
        return "Tier 4 - Low Priority"

    if score >= 75:
        return "Tier 1 - High Priority"

    elif score >= 60:
        return "Tier 2 - Priority"

    elif score >= 40:
        return "Tier 3 - Nurture"

    else:
        return "Tier 4 - Low Priority"


scoring_df[
    "Lead_Priority_Workflow"
] = scoring_df.apply(
    assign_priority,
    axis=1
)


# ------------------------------------------------------------
# 11. PROSPECT RANK
# ------------------------------------------------------------

scoring_df[
    "Prospect_Rank_Workflow"
] = (
    scoring_df[
        "Prospect_Score_Workflow"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 12. SCORE INTERPRETATION
# ------------------------------------------------------------

def score_interpretation(score):

    if score >= 75:
        return "High potential prospect"

    elif score >= 60:
        return "Priority prospect"

    elif score >= 40:
        return "Nurture prospect"

    return "Low priority prospect"


scoring_df[
    "Prospect_Score_Interpretation_Workflow"
] = scoring_df[
    "Prospect_Score_Workflow"
].apply(
    score_interpretation
)


# ------------------------------------------------------------
# 13. SUMMARY
# ------------------------------------------------------------

priority_summary = (
    scoring_df[
        "Lead_Priority_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Lead_Priority"
    )
    .reset_index(
        name="Company_Count"
    )
)

priority_summary["Percentage"] = (
    priority_summary["Company_Count"]
    / len(scoring_df)
    * 100
).round(2)


print("\nLead Priority Summary:")
display(priority_summary)


# ------------------------------------------------------------
# 14. SCORE SUMMARY
# ------------------------------------------------------------

score_summary = pd.DataFrame({
    "Metric": [
        "Prospect Score",
        "Company Fit",
        "Technology Fit",
        "Growth Intent",
        "Funding / Expansion",
        "Marketing Readiness",
        "Data Confidence"
    ],
    "Average": [
        scoring_df[
            "Prospect_Score_Workflow"
        ].mean(),
        scoring_df[
            "Company_Fit_Score_Workflow"
        ].mean(),
        scoring_df[
            "Technology_Fit_Score_Workflow"
        ].mean(),
        scoring_df[
            "Growth_Intent_Score_Workflow"
        ].mean(),
        scoring_df[
            "Funding_Expansion_Score_Workflow"
        ].mean(),
        scoring_df[
            "Marketing_Readiness_Score_Workflow"
        ].mean(),
        scoring_df[
            "Data_Confidence_Score_Workflow"
        ].mean()
    ]
}).round(2)

print("\nScoring Summary:")
display(score_summary)


# ------------------------------------------------------------
# 15. PRIORITY × READINESS VALIDATION
# ------------------------------------------------------------

priority_readiness_check = pd.crosstab(
    scoring_df[
        "Lead_Priority_Workflow"
    ],
    scoring_df[
        "Marketing_Readiness_Workflow"
    ]
)

print("\nPriority vs Marketing Readiness:")
display(priority_readiness_check)


# ------------------------------------------------------------
# 16. CHECK SCORE RANGE
# ------------------------------------------------------------

print("\nScore Range:")
print(
    "Minimum:",
    scoring_df[
        "Prospect_Score_Workflow"
    ].min()
)

print(
    "Maximum:",
    scoring_df[
        "Prospect_Score_Workflow"
    ].max()
)


# ------------------------------------------------------------
# 17. SAVE ANALYSIS REPORTS
# ------------------------------------------------------------

priority_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Lead_Priority_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

score_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Scoring_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

priority_readiness_check.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Priority_Readiness_Check.csv"
    ),
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 18. SAVE STEP OUTPUT
# ------------------------------------------------------------

step14_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_14_Scoring_Priority.csv"
)

scoring_df.to_csv(
    step14_output,
    index=False,
    encoding="utf-8-sig"
)


print("=" * 65)
print("STEP 14 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(scoring_df)
)

print(
    "Score range:",
    scoring_df["Prospect_Score_Workflow"].min(),
    "to",
    scoring_df["Prospect_Score_Workflow"].max()
)

print("\nOutput:")
print(step14_output)
```

```python
# ============================================================
# STEP 15 — CAMPAIGN INTELLIGENCE
# ============================================================

campaign_df = scoring_df.copy()

print("Rows entering Step 15:", len(campaign_df))


# ------------------------------------------------------------
# 1. HELPER
# ------------------------------------------------------------

def get_value(row, columns, default="Not Available"):

    for col in columns:

        if col in row.index:

            value = row[col]

            if pd.notna(value):

                text = str(value).strip()

                if text and text.lower() not in [
                    "nan",
                    "none",
                    "not available",
                    "n/a"
                ]:
                    return text

    return default


def contains_any(text, keywords):

    text = str(text).lower()

    return any(
        keyword.lower() in text
        for keyword in keywords
    )


# ------------------------------------------------------------
# 2. BUILD EVIDENCE TEXT
# ------------------------------------------------------------

evidence_columns = [
    "Observed_Information_Workflow",
    "Technology_Evidence_Text_Workflow",
    "AI_Evidence_Workflow",
    "Cloud_Data_Evidence_Workflow",
    "API_Platform_Evidence_Workflow",
    "Software_Development_Evidence_Workflow",
    "Hiring_Signals_Workflow",
    "Growth_Signals_Workflow",
    "Funding_Signals_Workflow",
    "Expansion_Signals_Workflow",
    "Marketing_Segment_Reason_Workflow"
]

available_evidence_columns = [
    col
    for col in evidence_columns
    if col in campaign_df.columns
]

if available_evidence_columns:

    campaign_df[
        "Campaign_Evidence_Text_Workflow"
    ] = (
        campaign_df[
            available_evidence_columns
        ]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
    )

else:

    campaign_df[
        "Campaign_Evidence_Text_Workflow"
    ] = ""


# ------------------------------------------------------------
# 3. TARGET PERSONA
# ------------------------------------------------------------

def campaign_persona(row):

    persona = get_value(
        row,
        [
            "Persona_Workflow",
            "Persona"
        ],
        "Needs Verification"
    )

    if persona not in [
        "Not Available",
        "Needs Verification",
        "nan"
    ]:
        return persona

    # Do NOT fabricate a persona.
    return "Needs Verification"


campaign_df[
    "Campaign_Target_Persona_Workflow"
] = campaign_df.apply(
    campaign_persona,
    axis=1
)


# ------------------------------------------------------------
# 4. CAMPAIGN SERVICE
# ------------------------------------------------------------

def campaign_service(row):

    primary = get_value(
        row,
        [
            "Primary_Service_Opportunity_Workflow"
        ]
    )

    if primary not in [
        "Not Available",
        "No Evidence Available"
    ]:
        return primary

    secondary = get_value(
        row,
        [
            "Secondary_Service_Opportunity_Workflow"
        ]
    )

    if secondary not in [
        "Not Available",
        "No Evidence Available"
    ]:
        return secondary

    return "Needs Verification"


campaign_df[
    "Campaign_Service_Workflow"
] = campaign_df.apply(
    campaign_service,
    axis=1
)


# ------------------------------------------------------------
# 5. IDENTIFY SIGNALS
# ------------------------------------------------------------

def signal_flags(row):

    text = row[
        "Campaign_Evidence_Text_Workflow"
    ]

    return {
        "AI": contains_any(
            text,
            [
                "ai",
                "artificial intelligence",
                "machine learning",
                "genai",
                "llm"
            ]
        ),

        "Cloud": contains_any(
            text,
            [
                "cloud",
                "aws",
                "azure",
                "gcp",
                "cloud data"
            ]
        ),

        "API": contains_any(
            text,
            [
                "api",
                "platform",
                "microservices",
                "graphql"
            ]
        ),

        "Software": contains_any(
            text,
            [
                "software development",
                "software",
                "application development"
            ]
        ),

        "Automation": contains_any(
            text,
            [
                "automation",
                "workflow automation"
            ]
        ),

        "Data": contains_any(
            text,
            [
                "data analytics",
                "data engineering",
                "analytics",
                "databricks",
                "snowflake"
            ]
        ),

        "Hiring": contains_any(
            text,
            [
                "hiring",
                "recruit",
                "job opening",
                "jobs"
            ]
        ),

        "Funding": contains_any(
            text,
            [
                "funding",
                "funded",
                "investment",
                "raised",
                "series a",
                "series b",
                "series c"
            ]
        ),

        "Growth": contains_any(
            text,
            [
                "growth",
                "expansion",
                "expanding",
                "scaling"
            ]
        )
    }


# ------------------------------------------------------------
# 6. CAMPAIGN ANGLE
# ------------------------------------------------------------

def recommended_campaign_angle(row):

    flags = signal_flags(row)

    service = row[
        "Campaign_Service_Workflow"
    ]

    segment = get_value(
        row,
        [
            "Marketing_Segment_Workflow"
        ]
    )

    priority = get_value(
        row,
        [
            "Lead_Priority_Workflow"
        ]
    )

    if flags["AI"]:

        return (
            "AI/ML modernization and GenAI solution "
            "discussion focused on practical business use cases."
        )

    if flags["Cloud"] and flags["Data"]:

        return (
            "Cloud and data engineering modernization "
            "focused on scalability, reliability and analytics."
        )

    if flags["API"]:

        return (
            "API/platform modernization and integration "
            "focused on scalable backend architecture."
        )

    if flags["Automation"]:

        return (
            "Workflow automation opportunity focused on "
            "reducing manual processes and improving efficiency."
        )

    if flags["Hiring"]:

        return (
            "Engineering capacity and delivery support "
            "through custom software or dedicated development resources."
        )

    if flags["Funding"] or flags["Growth"]:

        return (
            "Technology scaling support aligned with "
            "growth, expansion and product delivery needs."
        )

    if service != "Needs Verification":

        return (
            f"{service} opportunity based on observed "
            f"technology/business evidence."
        )

    if segment != "Not Available":

        return (
            f"Targeted outreach based on the observed "
            f"{segment} segment."
        )

    if priority == "Tier 1 - High Priority":

        return (
            "High-priority technology prospect requiring "
            "targeted business development outreach."
        )

    return (
        "General B2B technology outreach requiring "
        "additional qualification."
    )


campaign_df[
    "Recommended_Campaign_Angle_Workflow"
] = campaign_df.apply(
    recommended_campaign_angle,
    axis=1
)


# ------------------------------------------------------------
# 7. TARGETING REASON
# ------------------------------------------------------------

def campaign_targeting_reason(row):

    reasons = []

    priority = row[
        "Lead_Priority_Workflow"
    ]

    readiness = row[
        "Marketing_Readiness_Workflow"
    ]

    score = row[
        "Prospect_Score_Workflow"
    ]

    service = row[
        "Campaign_Service_Workflow"
    ]

    segment = get_value(
        row,
        [
            "Marketing_Segment_Workflow"
        ]
    )

    flags = signal_flags(row)

    if flags["AI"]:
        reasons.append(
            "Observed AI/ML evidence"
        )

    if flags["Cloud"]:
        reasons.append(
            "Observed Cloud evidence"
        )

    if flags["API"]:
        reasons.append(
            "Observed API/Platform evidence"
        )

    if flags["Software"]:
        reasons.append(
            "Observed Software Development evidence"
        )

    if flags["Data"]:
        reasons.append(
            "Observed Data/Analytics evidence"
        )

    if flags["Automation"]:
        reasons.append(
            "Observed Automation evidence"
        )

    if flags["Hiring"]:
        reasons.append(
            "Observed Hiring evidence"
        )

    if flags["Funding"]:
        reasons.append(
            "Observed Funding evidence"
        )

    if flags["Growth"]:
        reasons.append(
            "Observed Growth/Expansion evidence"
        )

    if service != "Needs Verification":
        reasons.append(
            f"Potential service opportunity: {service}"
        )

    if segment != "Not Available":
        reasons.append(
            f"Segment: {segment}"
        )

    reasons.append(
        f"Priority: {priority}"
    )

    reasons.append(
        f"Prospect score: {score}"
    )

    reasons.append(
        f"Marketing readiness: {readiness}"
    )

    return "; ".join(reasons)


campaign_df[
    "Campaign_Targeting_Reason_Workflow"
] = campaign_df.apply(
    campaign_targeting_reason,
    axis=1
)


# ------------------------------------------------------------
# 8. CAMPAIGN EVIDENCE TYPE
# ------------------------------------------------------------

def evidence_type(row):

    evidence = row[
        "Campaign_Evidence_Text_Workflow"
    ].strip()

    if not evidence:
        return "No Observed Evidence"

    return "Observed Evidence + Inferred Opportunity"


campaign_df[
    "Campaign_Evidence_Type_Workflow"
] = campaign_df.apply(
    evidence_type,
    axis=1
)


# ------------------------------------------------------------
# 9. CAMPAIGN PRIORITY
# ------------------------------------------------------------

campaign_df[
    "Campaign_Priority_Workflow"
] = campaign_df[
    "Lead_Priority_Workflow"
]


# ------------------------------------------------------------
# 10. CAMPAIGN STATUS
# ------------------------------------------------------------

def campaign_status(row):

    if row[
        "Marketing_Readiness_Workflow"
    ] == "DO NOT USE":

        return "DO NOT USE"

    if row[
        "Marketing_Readiness_Workflow"
    ] == "READY":

        return "READY FOR TARGETED CAMPAIGN"

    return "REQUIRES VERIFICATION"


campaign_df[
    "Campaign_Status_Workflow"
] = campaign_df.apply(
    campaign_status,
    axis=1
)


# ------------------------------------------------------------
# 11. FINAL CAMPAIGN INTELLIGENCE TABLE
# ------------------------------------------------------------

campaign_columns = [
    "Campaign_Target_Persona_Workflow",
    "Marketing_Segment_Workflow",
    "Campaign_Service_Workflow",
    "Recommended_Campaign_Angle_Workflow",
    "Campaign_Targeting_Reason_Workflow",
    "Campaign_Evidence_Type_Workflow",
    "Campaign_Priority_Workflow",
    "Campaign_Status_Workflow"
]

campaign_intelligence = campaign_df[
    campaign_columns
].copy()


# ------------------------------------------------------------
# 12. VALIDATION
# ------------------------------------------------------------

print("\nCampaign Intelligence Validation:")

for col in campaign_columns:

    missing = (
        campaign_df[col]
        .isna()
        .sum()
    )

    empty = (
        campaign_df[col]
        .astype(str)
        .str.strip()
        .eq("")
        .sum()
    )

    print(
        f"{col}: "
        f"Missing={missing}, "
        f"Empty={empty}"
    )


# ------------------------------------------------------------
# 13. SUMMARY BY CAMPAIGN SERVICE
# ------------------------------------------------------------

campaign_service_summary = (
    campaign_df[
        "Campaign_Service_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Campaign_Service"
    )
    .reset_index(
        name="Company_Count"
    )
)

print("\nCampaign Service Summary:")
display(campaign_service_summary)


# ------------------------------------------------------------
# 14. SUMMARY BY TARGET PERSONA
# ------------------------------------------------------------

campaign_persona_summary = (
    campaign_df[
        "Campaign_Target_Persona_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Target_Persona"
    )
    .reset_index(
        name="Company_Count"
    )
)

print("\nCampaign Persona Summary:")
display(campaign_persona_summary)


# ------------------------------------------------------------
# 15. SUMMARY BY CAMPAIGN STATUS
# ------------------------------------------------------------

campaign_status_summary = (
    campaign_df[
        "Campaign_Status_Workflow"
    ]
    .value_counts()
    .rename_axis(
        "Campaign_Status"
    )
    .reset_index(
        name="Company_Count"
    )
)

print("\nCampaign Status Summary:")
display(campaign_status_summary)


# ------------------------------------------------------------
# 16. SAVE REPORTS
# ------------------------------------------------------------

campaign_service_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Campaign_Service_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

campaign_persona_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Campaign_Persona_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

campaign_status_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Campaign_Status_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)

campaign_intelligence.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Campaign_Intelligence.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 17. SAVE STEP OUTPUT
# ------------------------------------------------------------

step15_output = os.path.join(
    ENRICHED_DIR,
    "Reusable_Workflow_Step_15_Campaign_Intelligence.csv"
)

campaign_df.to_csv(
    step15_output,
    index=False,
    encoding="utf-8-sig"
)


print("=" * 65)
print("STEP 15 COMPLETE")
print("=" * 65)

print(
    "Rows preserved:",
    len(campaign_df)
)

print(
    "Campaign intelligence fields created:",
    len(campaign_columns)
)

print("\nOutput:")
print(step15_output)
```

```python
# ============================================================
# STEP 17 — CAMPAIGN OUTCOME TRACKING
# ============================================================

tracking_df = final_df.copy()

print("Rows entering Step 17:", len(tracking_df))


# ------------------------------------------------------------
# 1. CAMPAIGN TRACKING FIELDS
# ------------------------------------------------------------

tracking_fields = {

    "Campaign_Name": "Not Started",

    "Email_Sent": 0,

    "Delivered": 0,

    "Opened": 0,

    "Clicked": 0,

    "Replied": 0,

    "Interested": 0,

    "Meeting_Booked": 0,

    "Requirement_Received": 0,

    "Proposal_Sent": 0,

    "Won": 0,

    "Lost": 0,

    "Follow_up_Status": "Not Started",

    "Follow_up_Date": pd.NaT
}


# ------------------------------------------------------------
# 2. ADD TRACKING COLUMNS
# ------------------------------------------------------------

for col, default_value in tracking_fields.items():

    if col not in tracking_df.columns:

        tracking_df[col] = default_value


# ------------------------------------------------------------
# 3. ENSURE NUMERIC CAMPAIGN METRICS
# ------------------------------------------------------------

campaign_numeric_fields = [
    "Email_Sent",
    "Delivered",
    "Opened",
    "Clicked",
    "Replied",
    "Interested",
    "Meeting_Booked",
    "Requirement_Received",
    "Proposal_Sent",
    "Won",
    "Lost"
]

for col in campaign_numeric_fields:

    tracking_df[col] = pd.to_numeric(
        tracking_df[col],
        errors="coerce"
    ).fillna(0).astype(int)


# ------------------------------------------------------------
# 4. FOLLOW-UP DATE
# ------------------------------------------------------------

tracking_df["Follow_up_Date"] = pd.to_datetime(
    tracking_df["Follow_up_Date"],
    errors="coerce"
)


# ------------------------------------------------------------
# 5. CAMPAIGN TRACKING STATUS
# ------------------------------------------------------------

def determine_followup_status(row):

    if row["Won"] > 0:
        return "Won"

    if row["Lost"] > 0:
        return "Lost"

    if row["Proposal_Sent"] > 0:
        return "Proposal Sent"

    if row["Requirement_Received"] > 0:
        return "Requirement Received"

    if row["Meeting_Booked"] > 0:
        return "Meeting Booked"

    if row["Interested"] > 0:
        return "Interested"

    if row["Replied"] > 0:
        return "Replied"

    if row["Clicked"] > 0:
        return "Clicked"

    if row["Opened"] > 0:
        return "Opened"

    if row["Email_Sent"] > 0:
        return "Sent"

    return "Not Started"


tracking_df[
    "Follow_up_Status"
] = tracking_df.apply(
    determine_followup_status,
    axis=1
)


# ------------------------------------------------------------
# 6. CAMPAIGN PERFORMANCE CALCULATIONS
# ------------------------------------------------------------

tracking_df["Delivery_Rate"] = np.where(
    tracking_df["Email_Sent"] > 0,
    (
        tracking_df["Delivered"]
        / tracking_df["Email_Sent"]
        * 100
    ).round(2),
    0
)

tracking_df["Open_Rate"] = np.where(
    tracking_df["Delivered"] > 0,
    (
        tracking_df["Opened"]
        / tracking_df["Delivered"]
        * 100
    ).round(2),
    0
)

tracking_df["Click_Rate"] = np.where(
    tracking_df["Delivered"] > 0,
    (
        tracking_df["Clicked"]
        / tracking_df["Delivered"]
        * 100
    ).round(2),
    0
)

tracking_df["Reply_Rate"] = np.where(
    tracking_df["Delivered"] > 0,
    (
        tracking_df["Replied"]
        / tracking_df["Delivered"]
        * 100
    ).round(2),
    0
)

tracking_df["Interest_Rate"] = np.where(
    tracking_df["Replied"] > 0,
    (
        tracking_df["Interested"]
        / tracking_df["Replied"]
        * 100
    ).round(2),
    0
)

tracking_df["Meeting_Rate"] = np.where(
    tracking_df["Delivered"] > 0,
    (
        tracking_df["Meeting_Booked"]
        / tracking_df["Delivered"]
        * 100
    ).round(2),
    0
)


# ------------------------------------------------------------
# 7. OUTCOME STAGE
# ------------------------------------------------------------

def determine_outcome_stage(row):

    if row["Won"] > 0:
        return "Won"

    if row["Lost"] > 0:
        return "Lost"

    if row["Proposal_Sent"] > 0:
        return "Proposal Sent"

    if row["Requirement_Received"] > 0:
        return "Requirement Received"

    if row["Meeting_Booked"] > 0:
        return "Meeting Booked"

    if row["Interested"] > 0:
        return "Interested"

    if row["Replied"] > 0:
        return "Replied"

    if row["Clicked"] > 0:
        return "Clicked"

    if row["Opened"] > 0:
        return "Opened"

    if row["Delivered"] > 0:
        return "Delivered"

    if row["Email_Sent"] > 0:
        return "Email Sent"

    return "Not Contacted"


tracking_df[
    "Campaign_Outcome_Stage"
] = tracking_df.apply(
    determine_outcome_stage,
    axis=1
)


# ------------------------------------------------------------
# 8. CAMPAIGN ELIGIBILITY
# ------------------------------------------------------------

def campaign_eligibility(row):

    readiness = row[
        "Marketing_Readiness"
    ]

    priority = row[
        "Lead_Priority"
    ]

    if readiness == "DO NOT USE":
        return "DO NOT USE"

    if readiness == "READY":

        if priority in [
            "Tier 1 - High Priority",
            "Tier 2 - Priority"
        ]:
            return "CAMPAIGN READY"

        return "NURTURE / QUALIFY"

    return "REQUIRES VERIFICATION"


tracking_df[
    "Campaign_Eligibility"
] = tracking_df.apply(
    campaign_eligibility,
    axis=1
)


# ------------------------------------------------------------
# 9. CAMPAIGN TRACKING SUMMARY
# ------------------------------------------------------------

tracking_summary = pd.DataFrame({

    "Metric": [
        "Total Companies",
        "Campaign Ready",
        "Requires Verification",
        "Do Not Use",
        "Emails Sent",
        "Delivered",
        "Opened",
        "Clicked",
        "Replied",
        "Interested",
        "Meetings Booked",
        "Requirements Received",
        "Proposals Sent",
        "Won",
        "Lost"
    ],

    "Value": [

        len(tracking_df),

        (
            tracking_df[
                "Campaign_Eligibility"
            ]
            .eq("CAMPAIGN READY")
            .sum()
        ),

        (
            tracking_df[
                "Campaign_Eligibility"
            ]
            .eq("REQUIRES VERIFICATION")
            .sum()
        ),

        (
            tracking_df[
                "Campaign_Eligibility"
            ]
            .eq("DO NOT USE")
            .sum()
        ),

        tracking_df["Email_Sent"].sum(),

        tracking_df["Delivered"].sum(),

        tracking_df["Opened"].sum(),

        tracking_df["Clicked"].sum(),

        tracking_df["Replied"].sum(),

        tracking_df["Interested"].sum(),

        tracking_df["Meeting_Booked"].sum(),

        tracking_df["Requirement_Received"].sum(),

        tracking_df["Proposal_Sent"].sum(),

        tracking_df["Won"].sum(),

        tracking_df["Lost"].sum()
    ]
})


print("\nCAMPAIGN TRACKING SUMMARY")
display(tracking_summary)


# ------------------------------------------------------------
# 10. VALIDATE TRACKING STRUCTURE
# ------------------------------------------------------------

required_tracking_fields = [
    "Campaign_Name",
    "Email_Sent",
    "Delivered",
    "Opened",
    "Clicked",
    "Replied",
    "Interested",
    "Meeting_Booked",
    "Requirement_Received",
    "Proposal_Sent",
    "Won",
    "Lost",
    "Follow_up_Status",
    "Follow_up_Date"
]

missing_tracking_fields = [
    col
    for col in required_tracking_fields
    if col not in tracking_df.columns
]


print("\nTracking Field Validation")

if not missing_tracking_fields:

    print(
        "PASS — All campaign tracking fields exist."
    )

else:

    print(
        "FAIL:",
        missing_tracking_fields
    )


# ------------------------------------------------------------
# 11. SAVE TRACKING SUMMARY
# ------------------------------------------------------------

tracking_summary.to_csv(
    os.path.join(
        ANALYSIS_DIR,
        "Reusable_Workflow_Campaign_Tracking_Summary.csv"
    ),
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 12. SAVE CAMPAIGN TRACKING DATASET
# ------------------------------------------------------------

campaign_tracking_path = os.path.join(
    FINAL_DIR,
    "NexziByte_Campaign_Outcome_Tracking.csv"
)

tracking_df.to_csv(
    campaign_tracking_path,
    index=False,
    encoding="utf-8-sig"
)


# ------------------------------------------------------------
# 13. FINAL OUTPUT
# ------------------------------------------------------------

print("=" * 70)
print("STEP 17 COMPLETE")
print("=" * 70)

print(
    "Rows:",
    len(tracking_df)
)

print(
    "Tracking fields:",
    len(required_tracking_fields)
)

print("\nCampaign Tracking File:")
print(campaign_tracking_path)
```

```python
# ============================================================
# STEP 19C — FIX SIGNAL, EVIDENCE & SCORING HANDOFF
# ============================================================

print("=" * 70)
print("FIXING SIGNAL / EVIDENCE / GROWTH / FUNDING HANDOFF")
print("=" * 70)

repair_df = final_df.copy()


# ------------------------------------------------------------
# 1. COPY SIGNAL FIELDS FROM THEIR ACTUAL WORKFLOW COLUMNS
# ------------------------------------------------------------

signal_mapping = {

    "AI_ML_Signals": [
        "AI_ML_Signals_Workflow",
        "AI/ML Signals",
        "AI_ML_Signals"
    ],

    "Cloud_Data_Signals": [
        "Cloud_Data_Signals_Workflow",
        "Cloud/Data Signals",
        "Cloud_Data_Signals"
    ],

    "API_Platform_Signals": [
        "API_Platform_Signals_Workflow",
        "API/Platform Signals",
        "API_Platform_Signals"
    ],

    "Software_Development_Signals": [
        "Software_Development_Signals_Workflow",
        "Software Development Signals",
        "Software_Development_Signals"
    ],

    "Automation_Workflow_Signals": [
        "Automation_Workflow_Signals_Workflow",
        "Automation / Workflow Signals",
        "Automation_Workflow_Signals"
    ],

    "Hiring_Signals": [
        "Hiring_Signals_Workflow",
        "Hiring Signals"
    ],

    "Growth_Signals": [
        "Growth_Signals_Workflow",
        "Growth Signals"
    ],

    "Expansion_Signals": [
        "Expansion_Signals_Workflow",
        "Expansion Signals"
    ],

    "Funding_Signals": [
        "Funding_Signals_Workflow",
        "Funding Signals"
    ],

    "Product_Launch_Signals": [
        "Product_Launch_Signals_Workflow",
        "Product Launch Signals"
    ],

    "Partnership_Signals": [
        "Partnership_Signals_Workflow",
        "Partnership Signals"
    ]
}


def first_existing_column(df, candidates):

    for col in candidates:

        if col in df.columns:
            return col

    return None


for final_col, candidates in signal_mapping.items():

    source_col = first_existing_column(
        campaign_df,
        candidates
    )

    if source_col:

        repair_df[final_col] = (
            campaign_df[source_col]
            .copy()
        )

        print(
            f"{final_col} <- {source_col}"
        )

    else:

        print(
            f"{final_col} -> No source available"
        )

        repair_df[final_col] = (
            "Not Available"
        )


# ------------------------------------------------------------
# 2. FIX OBSERVED INFORMATION
# ------------------------------------------------------------

observed_source_candidates = [
    "Observed_Information_Workflow",
    "Observed_Information",
    "Campaign_Evidence_Text_Workflow"
]

observed_source = first_existing_column(
    campaign_df,
    observed_source_candidates
)

if observed_source:

    repair_df[
        "Observed_Information"
    ] = (
        campaign_df[observed_source]
        .astype("string")
        .fillna("Not Available")
        .str.strip()
    )

else:

    repair_df[
        "Observed_Information"
    ] = "Not Available"


# ------------------------------------------------------------
# 3. REBUILD INFERRED OPPORTUNITY
# ------------------------------------------------------------

primary_service = (
    repair_df[
        "Primary_Service_Opportunity"
    ]
    .astype(str)
)

secondary_service = (
    repair_df[
        "Secondary_Service_Opportunity"
    ]
    .astype(str)
)


def build_inferred_opportunity(row):

    primary = str(
        row["Primary_Service_Opportunity"]
    )

    secondary = str(
        row["Secondary_Service_Opportunity"]
    )

    reason = str(
        row.get(
            "Service_Opportunity_Reason",
            "Not Available"
        )
    )

    if primary not in [
        "Not Available",
        "No Evidence Available",
        "nan"
    ]:

        result = (
            f"Potential NexziByte opportunity: "
            f"{primary}."
        )

        if secondary not in [
            "Not Available",
            "No Evidence Available",
            "nan"
        ]:

            result += (
                f" Secondary opportunity: {secondary}."
            )

        if reason not in [
            "Not Available",
            "nan"
        ]:

            result += (
                f" Basis: {reason}"
            )

        return result

    return "No evidence-based service opportunity identified."


repair_df[
    "Inferred_Opportunity"
] = repair_df.apply(
    build_inferred_opportunity,
    axis=1
)


# ------------------------------------------------------------
# 4. REBUILD GROWTH / FUNDING TEXT
# ------------------------------------------------------------

def combine_fields(row, fields):

    values = []

    for col in fields:

        if col not in row.index:
            continue

        value = row[col]

        if pd.isna(value):
            continue

        text = str(value).strip()

        if text and text.lower() not in [
            "not available",
            "nan",
            "none",
            ""
        ]:

            values.append(text)

    if values:
        return " | ".join(
            dict.fromkeys(values)
        )

    return "Not Available"


# If the original combined fields are available,
# use them as the authoritative source.

growth_combined = first_existing_column(
    campaign_df,
    [
        "Hiring_Growth_Signals",
        "Hiring/Growth Signals",
        "Hiring_Growth_Signals_Workflow"
    ]
)

funding_combined = first_existing_column(
    campaign_df,
    [
        "Funding_Expansion_Signals",
        "Funding/Expansion Signals",
        "Funding_Expansion_Signals_Workflow"
    ]
)


if growth_combined:

    growth_text = (
        campaign_df[growth_combined]
        .astype("string")
        .fillna("Not Available")
        .str.strip()
    )

else:

    growth_text = campaign_df.apply(
        lambda row: combine_fields(
            row,
            [
                "Hiring_Signals",
                "Hiring_Signals_Workflow",
                "Growth_Signals",
                "Growth_Signals_Workflow"
            ]
        ),
        axis=1
    )


if funding_combined:

    funding_text = (
        campaign_df[funding_combined]
        .astype("string")
        .fillna("Not Available")
        .str.strip()
    )

else:

    funding_text = campaign_df.apply(
        lambda row: combine_fields(
            row,
            [
                "Funding_Signals",
                "Funding_Signals_Workflow",
                "Expansion_Signals",
                "Expansion_Signals_Workflow"
            ]
        ),
        axis=1
    )


# ------------------------------------------------------------
# 5. CALCULATE GROWTH INTENT FROM ACTUAL EVIDENCE
# ------------------------------------------------------------

growth_keywords = [
    "hiring",
    "growth",
    "expansion",
    "expanding",
    "scaling",
    "recruit",
    "recruiting",
    "job opening",
    "job openings",
    "career",
    "new office",
    "new market"
]


def growth_score(text):

    text = str(text).lower()

    hits = sum(
        keyword in text
        for keyword in growth_keywords
    )

    return min(
        hits * 4,
        20
    )


repair_df[
    "Growth_Intent_Score"
] = growth_text.apply(
    growth_score
)


# ------------------------------------------------------------
# 6. FUNDING / EXPANSION SCORE
# ------------------------------------------------------------

funding_keywords = [
    "funding",
    "funded",
    "investment",
    "raised",
    "series a",
    "series b",
    "series c",
    "series d",
    "venture",
    "capital",
    "expansion",
    "expanding"
]


def funding_score(text):

    text = str(text).lower()

    hits = sum(
        keyword in text
        for keyword in funding_keywords
    )

    return min(
        hits * 3,
        15
    )


repair_df[
    "Funding_Expansion_Score"
] = funding_text.apply(
    funding_score
)


# ------------------------------------------------------------
# 7. REBUILD PROSPECT SCORE
# ------------------------------------------------------------

score_components = [
    "Company_Fit_Score",
    "Technology_Fit_Score",
    "Growth_Intent_Score",
    "Funding_Expansion_Score",
    "Marketing_Readiness_Score",
    "Data_Confidence_Score"
]

for col in score_components:

    repair_df[col] = pd.to_numeric(
        repair_df[col],
        errors="coerce"
    ).fillna(0)


repair_df[
    "Prospect_Score"
] = (
    repair_df[
        "Company_Fit_Score"
    ]
    +
    repair_df[
        "Technology_Fit_Score"
    ]
    +
    repair_df[
        "Growth_Intent_Score"
    ]
    +
    repair_df[
        "Funding_Expansion_Score"
    ]
    +
    repair_df[
        "Marketing_Readiness_Score"
    ]
    +
    repair_df[
        "Data_Confidence_Score"
    ]
).clip(
    0,
    100
).round(2)


# ------------------------------------------------------------
# 8. REBUILD LEAD PRIORITY
# ------------------------------------------------------------

def rebuild_priority(row):

    readiness = row[
        "Marketing_Readiness"
    ]

    score = row[
        "Prospect_Score"
    ]

    if readiness == "DO NOT USE":
        return "Tier 4 - Low Priority"

    if score >= 75:
        return "Tier 1 - High Priority"

    if score >= 60:
        return "Tier 2 - Priority"

    if score >= 40:
        return "Tier 3 - Nurture"

    return "Tier 4 - Low Priority"


repair_df[
    "Lead_Priority"
] = repair_df.apply(
    rebuild_priority,
    axis=1
)


# ------------------------------------------------------------
# 9. REBUILD RANK
# ------------------------------------------------------------

repair_df[
    "Prospect_Rank"
] = (
    repair_df[
        "Prospect_Score"
    ]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# ------------------------------------------------------------
# 10. VALIDATION
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("REPAIR VALIDATION")
print("=" * 70)


print("\nGrowth Intent Score:")
print(
    repair_df[
        "Growth_Intent_Score"
    ].describe()
)


print("\nFunding / Expansion Score:")
print(
    repair_df[
        "Funding_Expansion_Score"
    ].describe()
)


print("\nProspect Score:")
print(
    repair_df[
        "Prospect_Score"
    ].describe()
)


print("\nMarketing Readiness:")
display(
    repair_df[
        "Marketing_Readiness"
    ].value_counts()
)


print("\nLead Priority:")
display(
    repair_df[
        "Lead_Priority"
    ].value_counts()
)


print("\nObserved Information:")
print(
    repair_df[
        "Observed_Information"
    ]
    .astype(str)
    .ne("Not Available")
    .sum()
)


print("\nInferred Opportunity:")
print(
    repair_df[
        "Inferred_Opportunity"
    ]
    .astype(str)
    .ne("No evidence-based service opportunity identified.")
    .sum()
)


# ------------------------------------------------------------
# 11. SAVE REPAIRED MASTER
# ------------------------------------------------------------

final_df = repair_df.copy()

final_master_path = os.path.join(
    FINAL_DIR,
    "NexziByte_B2B_Lead_Intelligence_REUSABLE_FINAL_MASTER.csv"
)

final_df.to_csv(
    final_master_path,
    index=False,
    encoding="utf-8-sig"
)


print("\n" + "=" * 70)
print("STEP 19C COMPLETE")
print("=" * 70)

print(
    "Rows:",
    len(final_df)
)

print(
    "Columns:",
    len(final_df.columns)
)

print(
    "Saved:",
    final_master_path
)
```

```python
# ==============================================================
# STEP 19E — REPAIR GROWTH / FUNDING USING ACTUAL SOURCE COLUMNS
# ==============================================================

print("=" * 70)
print("REPAIRING GROWTH / FUNDING FROM ACTUAL SOURCE EVIDENCE")
print("=" * 70)

import numpy as np
import pandas as pd
import re


# --------------------------------------------------------------
# 1. SOURCE COLUMNS CONFIRMED FROM STEP 19D
# --------------------------------------------------------------

growth_source = campaign_df["Hiring_Growth_Signals"].fillna("").astype(str)
funding_source = campaign_df["Funding_Expansion_Signals"].fillna("").astype(str)

print("Growth source:", "Hiring_Growth_Signals")
print("Funding source:", "Funding_Expansion_Signals")


# --------------------------------------------------------------
# 2. NORMALIZE EVIDENCE
# --------------------------------------------------------------

def normalize_text(value):
    value = str(value).lower()
    value = value.replace("not available", "")
    value = value.replace("observed in supplied apollo data", "")
    return value.strip()


growth_text = growth_source.apply(normalize_text)
funding_text = funding_source.apply(normalize_text)


# --------------------------------------------------------------
# 3. GROWTH / HIRING / EXPANSION KEYWORDS
# --------------------------------------------------------------

growth_keywords = [
    "hiring",
    "hire",
    "recruit",
    "recruiting",
    "recruitment",
    "job opening",
    "job openings",
    "career",
    "careers",
    "talent acquisition",
    "new office",
    "new offices",
    "new market",
    "new markets",
    "expansion",
    "expanding",
    "expanded",
    "growth",
    "growing",
    "scaling",
    "scale",
    "market expansion"
]


# --------------------------------------------------------------
# 4. FUNDING / INVESTMENT KEYWORDS
# --------------------------------------------------------------

funding_keywords = [
    "funding",
    "funded",
    "investment",
    "invested",
    "raised",
    "series a",
    "series b",
    "series c",
    "series d",
    "series e",
    "venture",
    "venture capital",
    "capital",
    "private equity",
    "financing",
    "debt financing"
]


# --------------------------------------------------------------
# 5. SCORE FUNCTIONS
# --------------------------------------------------------------

def keyword_score(text, keywords, max_score):
    text = normalize_text(text)

    if not text:
        return 0.0

    matches = []

    for keyword in keywords:
        if keyword in text:
            matches.append(keyword)

    matches = list(dict.fromkeys(matches))

    if not matches:
        return 0.0

    # Evidence strength:
    # 1 match  = 5 points
    # 2 matches = 10 points
    # 3 matches = 15 points
    # 4+ matches = max
    score = min(len(matches) * 5, max_score)

    return float(score)


growth_scores = growth_text.apply(
    lambda x: keyword_score(
        x,
        growth_keywords,
        20
    )
)

funding_scores = funding_text.apply(
    lambda x: keyword_score(
        x,
        funding_keywords,
        15
    )
)


# --------------------------------------------------------------
# 6. WRITE SCORES INTO FINAL DATA
# --------------------------------------------------------------

final_df["Growth_Intent_Score"] = growth_scores
final_df["Funding_Expansion_Score"] = funding_scores


# Keep workflow versions synchronized
final_df["Growth_Intent_Score_Workflow"] = growth_scores
final_df["Funding_Expansion_Score_Workflow"] = funding_scores


# --------------------------------------------------------------
# 7. ADD EXPLICIT EVIDENCE COLUMNS
# --------------------------------------------------------------

final_df["Growth_Funding_Evidence_Type"] = np.where(
    (growth_scores > 0) | (funding_scores > 0),
    "Observed Source Evidence",
    "No Observed Evidence Available"
)

final_df["Growth_Signal_Source"] = growth_source
final_df["Funding_Expansion_Signal_Source"] = funding_source


# --------------------------------------------------------------
# 8. REBUILD PROSPECT SCORE
# --------------------------------------------------------------

score_columns = [
    "Company_Fit_Score",
    "Technology_Fit_Score",
    "Growth_Intent_Score",
    "Funding_Expansion_Score",
    "Marketing_Readiness_Score",
    "Data_Confidence_Score"
]

for col in score_columns:
    final_df[col] = pd.to_numeric(
        final_df[col],
        errors="coerce"
    ).fillna(0)

final_df["Prospect_Score"] = (
    final_df["Company_Fit_Score"]
    + final_df["Technology_Fit_Score"]
    + final_df["Growth_Intent_Score"]
    + final_df["Funding_Expansion_Score"]
    + final_df["Marketing_Readiness_Score"]
    + final_df["Data_Confidence_Score"]
).clip(0, 100)


# --------------------------------------------------------------
# 9. REBUILD PRIORITY
# --------------------------------------------------------------

def assign_priority(score, readiness):
    if readiness == "DO NOT USE":
        return "Tier 4 - Low Priority"

    if score >= 75:
        return "Tier 1 - High Priority"

    if score >= 60:
        return "Tier 2 - Priority"

    if score >= 40:
        return "Tier 3 - Nurture"

    return "Tier 4 - Low Priority"


final_df["Lead_Priority"] = final_df.apply(
    lambda row: assign_priority(
        row["Prospect_Score"],
        row["Marketing_Readiness"]
    ),
    axis=1
)


# --------------------------------------------------------------
# 10. REBUILD RANK
# --------------------------------------------------------------

final_df["Prospect_Rank"] = (
    final_df["Prospect_Score"]
    .rank(
        ascending=False,
        method="dense"
    )
    .astype(int)
)


# --------------------------------------------------------------
# 11. VALIDATION
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("GROWTH INTENT SCORE")
print("=" * 70)

print(final_df["Growth_Intent_Score"].describe())

print("\nNon-zero Growth Scores:")
print(
    (final_df["Growth_Intent_Score"] > 0).sum()
)


print("\n" + "=" * 70)
print("FUNDING / EXPANSION SCORE")
print("=" * 70)

print(final_df["Funding_Expansion_Score"].describe())

print("\nNon-zero Funding Scores:")
print(
    (final_df["Funding_Expansion_Score"] > 0).sum()
)


print("\n" + "=" * 70)
print("PROSPECT SCORE")
print("=" * 70)

print(final_df["Prospect_Score"].describe())


print("\n" + "=" * 70)
print("LEAD PRIORITY")
print("=" * 70)

print(
    final_df["Lead_Priority"].value_counts()
)


print("\n" + "=" * 70)
print("EVIDENCE STATUS")
print("=" * 70)

print(
    final_df["Growth_Funding_Evidence_Type"].value_counts()
)


# --------------------------------------------------------------
# 12. EXPORT
# --------------------------------------------------------------

output_path = (
    PROJECT_ROOT
    / "05_Final_Output"
    / "NexziByte_B2B_Lead_Intelligence_REUSABLE_FINAL_MASTER.csv"
)

final_df.to_csv(
    output_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 70)
print("STEP 19E COMPLETE")
print("=" * 70)

print("Rows:", len(final_df))
print("Columns:", len(final_df.columns))
print("Saved:", output_path)
```

```python
# ==============================================================
# STEP 21 — FINAL MASTER STRUCTURE REPAIR
# ==============================================================

import pandas as pd
import numpy as np
from pathlib import Path

print("=" * 70)
print("STEP 21 — FINAL MASTER STRUCTURE REPAIR")
print("=" * 70)

df = validation_df.copy()

# --------------------------------------------------------------
# 1. FIND COMPANY NAME SOURCE
# --------------------------------------------------------------

print("\nSearching for company-name source...")

company_candidates = [
    "Company Name",
    "Company_Name",
    "Company",
    "Organization Name",
    "Organization_Name",
    "Account Name",
    "Account_Name",
    "Name",
    "company_name",
    "company"
]

company_source = None

for col in company_candidates:
    if col in df.columns:
        company_source = col
        break

# If not found, search column names containing company
if company_source is None:

    company_matches = [
        c for c in df.columns
        if "company" in str(c).lower()
    ]

    print("Company-related columns found:")
    print(company_matches)

    if company_matches:
        company_source = company_matches[0]

print("Selected company source:", company_source)

if company_source is None:
    raise ValueError(
        "No company-name source column could be identified."
    )

# Create standardized Company field
df["Company"] = (
    df[company_source]
    .fillna("")
    .astype(str)
    .str.strip()
)

print(
    "Blank Company values:",
    (df["Company"] == "").sum()
)


# --------------------------------------------------------------
# 2. REBUILD OBSERVED INFORMATION
# --------------------------------------------------------------

print("\nRebuilding Observed_Information...")

observed_candidates = [
    "Observed_Information_Workflow",
    "Observed Information",
    "Observed_Information",
    "Campaign_Evidence_Text_Workflow",
    "Campaign_Evidence_Text",
    "Technology_Evidence_Text_Workflow",
    "Technology_Evidence_Text",
    "Signal_Status_Workflow"
]

observed_source = None

for col in observed_candidates:
    if col in df.columns:
        observed_source = col
        break

print("Observed source:", observed_source)

if observed_source:

    observed_text = (
        df[observed_source]
        .fillna("")
        .astype(str)
        .str.strip()
    )

    df["Observed_Information"] = (
        observed_text != ""
    )

else:

    # Fallback: derive from actual evidence fields
    evidence_columns = [
        "AI_ML_Signals",
        "Cloud_Data_Signals",
        "API_Platform_Signals",
        "Software_Development_Signals",
        "Hiring_Growth_Signals",
        "Funding_Expansion_Signals",
        "Automation_Workflow_Signals",
        "Product_Launch_Signals",
        "Partnership_Signals"
    ]

    available_evidence = [
        c for c in evidence_columns
        if c in df.columns
    ]

    evidence_text = (
        df[available_evidence]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
    )

    df["Observed_Information"] = (
        evidence_text
        .str.lower()
        .str.contains(
            "observed|evidence|available",
            regex=True,
            na=False
        )
    )


# --------------------------------------------------------------
# 3. REBUILD INFERRED OPPORTUNITY
# --------------------------------------------------------------

print("\nRebuilding Inferred_Opportunity...")

service_cols = [
    "Primary_Service_Opportunity",
    "Secondary_Service_Opportunity"
]

available_service_cols = [
    c for c in service_cols
    if c in df.columns
]

if available_service_cols:

    service_text = (
        df[available_service_cols]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
        .str.lower()
    )

    inferred_keywords = (
        "opportunity|potential|recommended|"
        "service|inferred"
    )

    df["Inferred_Opportunity"] = (
        service_text
        .str.contains(
            inferred_keywords,
            regex=True,
            na=False
        )
    )

else:

    df["Inferred_Opportunity"] = False


# --------------------------------------------------------------
# 4. ADD EXPLICIT EVIDENCE TYPE
# --------------------------------------------------------------

print("\nCreating evidence classification...")

df["Evidence_Type"] = np.select(
    [
        df["Observed_Information"].eq(True)
        & df["Inferred_Opportunity"].eq(True),

        df["Observed_Information"].eq(True)
        & df["Inferred_Opportunity"].eq(False),

        df["Observed_Information"].eq(False)
        & df["Inferred_Opportunity"].eq(True)
    ],
    [
        "Observed + Inferred",
        "Observed Evidence",
        "Inferred Opportunity"
    ],
    default="No Evidence Available"
)


# --------------------------------------------------------------
# 5. FUNDING EVIDENCE
# --------------------------------------------------------------

print("\nCreating explicit Funding_Evidence flag...")

funding_metadata = [
    c for c in [
        "Total_Funding",
        "Latest_Funding",
        "Latest_Funding_Amount"
    ]
    if c in df.columns
]

if funding_metadata:

    funding_numeric_or_text = (
        df[funding_metadata]
        .fillna("")
        .astype(str)
        .agg(" | ".join, axis=1)
        .str.strip()
    )

    df["Funding_Evidence"] = (
        funding_numeric_or_text
        .str.replace(
            "nan",
            "",
            regex=False
        )
        .str.strip()
        .ne("")
    )

else:

    df["Funding_Evidence"] = False


# --------------------------------------------------------------
# 6. EXPLICIT GROWTH EVIDENCE
# --------------------------------------------------------------

print("\nCreating explicit Growth_Evidence flag...")

growth_source_candidates = [
    "Hiring_Growth_Signals",
    "Hiring_Growth_Signals_Workflow",
    "Hiring/Growth Signals"
]

growth_source = None

for col in growth_source_candidates:
    if col in df.columns:
        growth_source = col
        break

if growth_source:

    growth_text = (
        df[growth_source]
        .fillna("")
        .astype(str)
        .str.lower()
        .str.strip()
    )

    df["Growth_Evidence"] = (
        growth_text
        .str.contains(
            "observed",
            na=False
        )
        &
        ~growth_text.eq("not available")
    )

else:

    df["Growth_Evidence"] = False


# --------------------------------------------------------------
# 7. REBUILD PROSPECT SCORE
# --------------------------------------------------------------

score_cols = [
    "Company_Fit_Score",
    "Technology_Fit_Score",
    "Growth_Intent_Score",
    "Funding_Expansion_Score",
    "Marketing_Readiness_Score",
    "Data_Confidence_Score"
]

for col in score_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(
            df[col],
            errors="coerce"
        ).fillna(0)

df["Prospect_Score"] = (
    df[score_cols]
    .sum(axis=1)
    .clip(0, 100)
)


# --------------------------------------------------------------
# 8. REBUILD PRIORITY
# --------------------------------------------------------------

df["Lead_Priority"] = np.select(
    [
        df["Marketing_Readiness"].eq("DO NOT USE"),
        df["Prospect_Score"] >= 75,
        df["Prospect_Score"] >= 60,
        df["Prospect_Score"] >= 40
    ],
    [
        "Tier 4 - Low Priority",
        "Tier 1 - High Priority",
        "Tier 2 - Priority",
        "Tier 3 - Nurture"
    ],
    default="Tier 4 - Low Priority"
)

df["Prospect_Rank"] = (
    df["Prospect_Score"]
    .rank(
        method="dense",
        ascending=False
    )
    .astype(int)
)


# --------------------------------------------------------------
# 9. SAVE REPAIRED FINAL MASTER
# --------------------------------------------------------------

final_path = (
    Path(PROJECT_ROOT)
    / "05_Final_Output"
    / "NexziByte_B2B_Lead_Intelligence_REUSABLE_FINAL_MASTER.csv"
)

df.to_csv(
    final_path,
    index=False,
    encoding="utf-8-sig"
)

print("\n" + "=" * 70)
print("STEP 21 COMPLETE")
print("=" * 70)

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Company source:", company_source)
print(
    "Blank companies:",
    (df["Company"] == "").sum()
)

print(
    "Observed Information:",
    df["Observed_Information"].value_counts().to_dict()
)

print(
    "Inferred Opportunity:",
    df["Inferred_Opportunity"].value_counts().to_dict()
)

print(
    "Funding Evidence:",
    df["Funding_Evidence"].value_counts().to_dict()
)

print(
    "Growth Evidence:",
    df["Growth_Evidence"].value_counts().to_dict()
)

print("Saved:", final_path)
print("File exists:", final_path.exists())
```

```python
# ==============================================================
# STEP 25 — FINAL VALIDATION V3
# ==============================================================

from pathlib import Path
import pandas as pd
import numpy as np

print("=" * 70)
print("STEP 25 — FINAL VALIDATION V3")
print("=" * 70)

# --------------------------------------------------------------
# LOAD FINAL MASTER
# --------------------------------------------------------------

final_path = (
    Path(PROJECT_ROOT)
    / "05_Final_Output"
    / "NexziByte_B2B_Lead_Intelligence_REUSABLE_FINAL_MASTER.csv"
)

final_df_v3 = pd.read_csv(
    final_path,
    encoding="utf-8-sig",
    low_memory=False
)

print("\nRows:", len(final_df_v3))
print("Columns:", len(final_df_v3.columns))


# --------------------------------------------------------------
# REQUIRED COLUMNS
# --------------------------------------------------------------

required_columns = [
    "Company",
    "Company_Name",
    "Company_Country",
    "Company_City",
    "Company_State",
    "Marketing_Readiness",
    "Company_Fit_Score",
    "Technology_Fit_Score",
    "Growth_Intent_Score",
    "Funding_Expansion_Score",
    "Marketing_Readiness_Score",
    "Data_Confidence_Score",
    "Prospect_Score",
    "Lead_Priority",
    "Prospect_Rank",
    "Primary_Service_Opportunity",
    "Secondary_Service_Opportunity",
    "Observed_Information",
    "Inferred_Opportunity",
    "Evidence_Type",
    "Funding_Evidence",
    "Growth_Evidence"
]

missing_columns = [
    c for c in required_columns
    if c not in final_df_v3.columns
]

print("\n" + "=" * 70)
print("REQUIRED COLUMNS")
print("=" * 70)

if missing_columns:
    print("FAIL")
    print(missing_columns)
else:
    print("PASS — All required columns present")


# --------------------------------------------------------------
# COMPANY VALIDATION
# --------------------------------------------------------------

company = (
    final_df_v3["Company"]
    .fillna("")
    .astype(str)
    .str.strip()
)

blank_company = company.eq("").sum()
unique_companies = company.replace("", np.nan).nunique()

print("\n" + "=" * 70)
print("COMPANY VALIDATION")
print("=" * 70)

print("Total rows:", len(final_df_v3))
print("Blank companies:", blank_company)
print("Unique companies:", unique_companies)


# --------------------------------------------------------------
# DUPLICATE VALIDATION
# --------------------------------------------------------------

exact_duplicates = final_df_v3.duplicated().sum()

print("\n" + "=" * 70)
print("DUPLICATE VALIDATION")
print("=" * 70)

print("Exact duplicate rows:", exact_duplicates)


# --------------------------------------------------------------
# SCORE VALIDATION
# --------------------------------------------------------------

score_columns = [
    "Company_Fit_Score",
    "Technology_Fit_Score",
    "Growth_Intent_Score",
    "Funding_Expansion_Score",
    "Marketing_Readiness_Score",
    "Data_Confidence_Score",
    "Prospect_Score"
]

print("\n" + "=" * 70)
print("SCORE VALIDATION")
print("=" * 70)

score_missing = {}

for col in score_columns:

    final_df_v3[col] = pd.to_numeric(
        final_df_v3[col],
        errors="coerce"
    )

    score_missing[col] = final_df_v3[col].isna().sum()

    print(
        f"{col}: "
        f"min={final_df_v3[col].min()} | "
        f"max={final_df_v3[col].max()} | "
        f"missing={final_df_v3[col].isna().sum()}"
    )


# --------------------------------------------------------------
# SCORE RECONCILIATION
# --------------------------------------------------------------

calculated_score = (
    final_df_v3[
        [
            "Company_Fit_Score",
            "Technology_Fit_Score",
            "Growth_Intent_Score",
            "Funding_Expansion_Score",
            "Marketing_Readiness_Score",
            "Data_Confidence_Score"
        ]
    ]
    .fillna(0)
    .sum(axis=1)
    .clip(0, 100)
)

score_difference = (
    final_df_v3["Prospect_Score"]
    - calculated_score
).abs()

score_mismatches = (
    score_difference > 0.01
).sum()

print("\n" + "=" * 70)
print("PROSPECT SCORE RECONCILIATION")
print("=" * 70)

print("Score mismatches:", score_mismatches)
print("Maximum difference:", score_difference.max())


# --------------------------------------------------------------
# READINESS VALIDATION
# --------------------------------------------------------------

valid_readiness = [
    "READY",
    "NEEDS VERIFICATION",
    "DO NOT USE"
]

invalid_readiness = (
    ~final_df_v3["Marketing_Readiness"]
    .isin(valid_readiness)
).sum()

print("\n" + "=" * 70)
print("MARKETING READINESS")
print("=" * 70)

print(
    final_df_v3["Marketing_Readiness"]
    .value_counts()
)

print("Invalid values:", invalid_readiness)


# --------------------------------------------------------------
# PRIORITY VALIDATION
# --------------------------------------------------------------

valid_priority = [
    "Tier 1 - High Priority",
    "Tier 2 - Priority",
    "Tier 3 - Nurture",
    "Tier 4 - Low Priority"
]

invalid_priority = (
    ~final_df_v3["Lead_Priority"]
    .isin(valid_priority)
).sum()

print("\n" + "=" * 70)
print("LEAD PRIORITY")
print("=" * 70)

print(
    final_df_v3["Lead_Priority"]
    .value_counts()
)

print("Invalid values:", invalid_priority)


# --------------------------------------------------------------
# EVIDENCE VALIDATION
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("EVIDENCE CLASSIFICATION")
print("=" * 70)

print(
    "Observed Information:"
)

print(
    final_df_v3["Observed_Information"]
    .value_counts(dropna=False)
)

print(
    "\nInferred Opportunity:"
)

print(
    final_df_v3["Inferred_Opportunity"]
    .value_counts(dropna=False)
)

print(
    "\nEvidence Type:"
)

print(
    final_df_v3["Evidence_Type"]
    .value_counts(dropna=False)
)


# --------------------------------------------------------------
# BUSINESS SIGNAL VALIDATION
# --------------------------------------------------------------

print("\n" + "=" * 70)
print("BUSINESS SIGNAL VALIDATION")
print("=" * 70)

print(
    "Funding Evidence:"
)

print(
    final_df_v3["Funding_Evidence"]
    .value_counts(dropna=False)
)

print(
    "\nGrowth Evidence:"
)

print(
    final_df_v3["Growth_Evidence"]
    .value_counts(dropna=False)
)

print(
    "\nGrowth Intent Score:"
)

print(
    final_df_v3["Growth_Intent_Score"]
    .value_counts()
    .sort_index()
)

print(
    "\nFunding / Expansion Score:"
)

print(
    final_df_v3["Funding_Expansion_Score"]
    .value_counts()
    .sort_index()
)


# --------------------------------------------------------------
# SERVICE VALIDATION
# --------------------------------------------------------------

primary_missing = (
    final_df_v3["Primary_Service_Opportunity"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
).sum()

secondary_missing = (
    final_df_v3["Secondary_Service_Opportunity"]
    .fillna("")
    .astype(str)
    .str.strip()
    .eq("")
).sum()

print("\n" + "=" * 70)
print("SERVICE VALIDATION")
print("=" * 70)

print("Primary service missing:", primary_missing)
print("Secondary service missing:", secondary_missing)

print("\nPrimary service distribution:")
print(
    final_df_v3["Primary_Service_Opportunity"]
    .value_counts()
)


# --------------------------------------------------------------
# FINAL CHECKLIST
# --------------------------------------------------------------

checks = {

    "Rows > 0":
        len(final_df_v3) > 0,

    "Required columns present":
        len(missing_columns) == 0,

    "Blank companies = 0":
        blank_company == 0,

    "Unique companies > 1":
        unique_companies > 1,

    "Exact duplicates = 0":
        exact_duplicates == 0,

    "Score fields complete":
        all(v == 0 for v in score_missing.values()),

    "Prospect score reconciles":
        score_mismatches == 0,

    "Marketing readiness valid":
        invalid_readiness == 0,

    "Lead priority valid":
        invalid_priority == 0,

    "Primary service complete":
        primary_missing == 0,

    "Secondary service complete":
        secondary_missing == 0
}

print("\n" + "=" * 70)
print("FINAL MASTER CHECKLIST")
print("=" * 70)

for check, result in checks.items():

    print(
        f"{'PASS' if result else 'FAIL'} | {check}"
    )

overall_pass = all(checks.values())

print("\n" + "=" * 70)

if overall_pass:

    print("FINAL MASTER STATUS: PASS")
    print("READY FOR POWER BI")
    print("FINAL MASTER IS VALIDATED")

else:

    print("FINAL MASTER STATUS: REVIEW REQUIRED")
    print("DO NOT LOAD INTO POWER BI YET")

print("=" * 70)
```
