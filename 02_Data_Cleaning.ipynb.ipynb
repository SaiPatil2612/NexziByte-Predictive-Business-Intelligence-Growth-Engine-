```python
import re

def normalize_url(value):
    if pd.isna(value):
        return ""
    
    value = str(value).strip().lower()
    value = re.sub(r"^https?://", "", value)
    value = re.sub(r"^www\.", "", value)
    value = value.split("/")[0]
    value = value.rstrip(".")
    
    return value

clean_df["Website_Normalized"] = clean_df["Website"].apply(normalize_url)
```

```python
def normalize_linkedin(value):
    if pd.isna(value):
        return ""
    
    value = str(value).strip().lower()
    value = re.sub(r"^https?://", "", value)
    value = re.sub(r"^www\.", "", value)
    value = value.rstrip("/")
    
    return value

clean_df["LinkedIn_Normalized"] = (
    clean_df["Company Linkedin Url"]
    .apply(normalize_linkedin)
)
```

```python
# Identify records with the same normalized LinkedIn OR Website

duplicate_candidates = clean_df[
    (
        (clean_df["LinkedIn_Normalized"] != "") &
        (clean_df["LinkedIn_Normalized"].duplicated(keep=False))
    )
    |
    (
        (clean_df["Website_Normalized"] != "") &
        (clean_df["Website_Normalized"].duplicated(keep=False))
    )
].copy()

print("Duplicate candidate records:", len(duplicate_candidates))
```

```python
def duplicate_reason(row):
    linkedin = row["LinkedIn_Normalized"]
    website = row["Website_Normalized"]
    
    linkedin_count = (
        clean_df["LinkedIn_Normalized"] == linkedin
    ).sum() if linkedin else 0
    
    website_count = (
        clean_df["Website_Normalized"] == website
    ).sum() if website else 0
    
    reasons = []
    
    if linkedin and linkedin_count > 1:
        reasons.append("Same LinkedIn")
    
    if website and website_count > 1:
        reasons.append("Same Website")
    
    return " + ".join(reasons)


duplicate_candidates["Duplicate_Reason"] = (
    duplicate_candidates.apply(duplicate_reason, axis=1)
)
```

```python
duplicate_candidates[
    [
        "Company Name",
        "Website",
        "Company Linkedin Url",
        "Contact",
        "Apollo Account Id",
        "Apollo Record Id",
        "Duplicate_Reason"
    ]
].sort_values(
    ["Duplicate_Reason", "Company Name"]
)
```

```python
# Count occurrences of each normalized identity
linkedin_counts = clean_df["LinkedIn_Normalized"].value_counts()
website_counts = clean_df["Website_Normalized"].value_counts()

# Add duplicate decision
def classify_duplicate(row):
    linkedin = row["LinkedIn_Normalized"]
    website = row["Website_Normalized"]

    same_linkedin = (
        linkedin != "" and linkedin_counts.get(linkedin, 0) > 1
    )

    same_website = (
        website != "" and website_counts.get(website, 0) > 1
    )

    if same_linkedin and same_website:
        return "Confirmed Duplicate"

    elif same_linkedin or same_website:
        return "Review Required"

    else:
        return "Unique"


duplicate_candidates["Duplicate_Decision"] = (
    duplicate_candidates.apply(classify_duplicate, axis=1)
)

duplicate_candidates[
    [
        "Company Name",
        "Website",
        "Company Linkedin Url",
        "Contact",
        "Apollo Account Id",
        "Apollo Record Id",
        "Duplicate_Reason",
        "Duplicate_Decision"
    ]
]
```

```python
# Text fields to standardize
text_columns = [
    "Company Name",
    "Company Country",
    "Company City",
    "Company State",
    "Company Address",
    "Industry",
    "Contact",
    "Keywords",
    "Technology Areas"
]

def clean_text(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Convert empty strings to missing
    if value == "":
        return np.nan

    # Remove repeated spaces
    value = re.sub(r"\s+", " ", value)

    return value


for col in text_columns:
    clean_df[col] = clean_df[col].apply(clean_text)

print("Text standardization completed.")
```

```python
def clean_url(value):
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    if value == "":
        return np.nan

    value = value.lower()

    # Remove spaces
    value = value.replace(" ", "")

    return value


url_columns = [
    "Website",
    "Company Linkedin Url",
    "Facebook Url"
]

for col in url_columns:
    clean_df[col] = clean_df[col].apply(clean_url)

print("URL standardization completed.")
```

```python
import re

def clean_technology_areas(value):
    if pd.isna(value):
        return pd.NA

    # Split comma-separated technologies
    technologies = str(value).split(",")

    # Remove whitespace
    technologies = [
        tech.strip()
        for tech in technologies
        if tech.strip()
    ]

    # Remove exact duplicates while preserving original order
    technologies = list(dict.fromkeys(technologies))

    if not technologies:
        return pd.NA

    return ", ".join(technologies)


clean_df["Technology Areas Cleaned"] = (
    clean_df["Technology Areas"]
    .apply(clean_technology_areas)
)

print("Technology Areas cleaned.")
print(
    "Missing cleaned technology values:",
    clean_df["Technology Areas Cleaned"].isna().sum()
)
```

```python
def marketing_readiness_v2(row):

    signal_count = row["Observed Signal Count"]

    technology_available = pd.notna(
        row["Technology Areas Cleaned"]
    )

    funding_available = (
        pd.notna(row["Total Funding"]) or
        pd.notna(row["Latest Funding Amount"]) or
        pd.notna(row["Last Raised At"])
    )

    hiring_available = row["Hiring/Growth Signals Available"]

    # Strong evidence
    if (
        signal_count >= 5
        and technology_available
        and (funding_available or hiring_available)
    ):
        return "High Potential"

    # Moderate evidence
    if signal_count >= 4 and technology_available:
        return "Medium Potential"

    # Some evidence
    if signal_count >= 2 or technology_available:
        return "Needs Validation"

    return "Insufficient Evidence"


clean_df["Marketing Readiness V2"] = clean_df.apply(
    marketing_readiness_v2,
    axis=1
)

print(
    clean_df["Marketing Readiness V2"].value_counts()
)
```

```python
def rebuild_data_quality_status(row):

    issues = []

    # Location: actual missing location data
    if row["Location_Data_Status"] != "Complete":
        issues.append("Location")

    # Industry: only actual review/corruption
    if row["Industry_Review_Flag"] in [
        "Review",
        "Structural Corruption"
    ]:
        issues.append("Industry")

    # Employees: only unresolved verification
    if row["Employee_Data_Status"] == "Needs Verification":
        issues.append("Employees")

    # Marketing: actual review/missing data
    if row["Marketing_Data_Status"] in [
        "Requires Review",
        "Missing Marketing Data"
    ]:
        issues.append("Marketing")

    # Structural corruption
    if row["Industry_Review_Flag"] == "Structural Corruption":
        return "Needs Manual Review"

    if len(issues) >= 2:
        return "Needs Manual Review"

    if len(issues) == 1:
        return "Review Required"

    return "Clean"


clean_df["Data Quality Status V2"] = clean_df.apply(
    rebuild_data_quality_status,
    axis=1
)

print(
    clean_df["Data Quality Status V2"].value_counts()
)
```

```python
# ============================================================
# FIX CONTROLLED TECHNOLOGY EXTRACTION — ROW BY ROW
# ============================================================

import re
import pandas as pd

# Use ONLY each company's own technology field
technology_source = (
    clean_df["Technology Areas Cleaned"]
    .fillna("")
    .astype(str)
    .str.lower()
)

# Controlled technology dictionaries
ai_terms = [
    "openai",
    "tensorflow",
    "pytorch",
    "machine learning",
    "deep learning",
    "llm",
    "artificial intelligence",
    "generative ai",
    "computer vision",
    "nlp",
]

cloud_terms = [
    "amazon ec2",
    "amazon s3",
    "aws cloud",
    "microsoft azure",
    "azure",
    "google cloud",
    "snowflake",
    "databricks",
    "bigquery",
    "digitalocean",
    "cloudflare",
]

platform_terms = [
    "salesforce",
    "salesforce sales cloud",
    "servicenow",
    "workday",
    "sap",
    "oracle",
    "stripe",
    "twilio",
    "zoho",
    "hubspot",
]

dev_terms = [
    "github",
    "gitlab",
    "jenkins",
    "docker",
    "kubernetes",
    "react",
    "node.js",
    "nodejs",
    "angular",
    "vue.js",
    "python",
    "java",
    "javascript",
    "typescript",
]

def extract_terms(text, terms):
    """
    Return only terms actually present in THIS company's
    technology text.
    """
    text = str(text).lower()

    found = []

    for term in terms:
        pattern = r"(?<![a-z0-9])" + re.escape(term.lower()) + r"(?![a-z0-9])"

        if re.search(pattern, text):
            found.append(term)

    return found


# Extract independently for every row
clean_df["High Value AI Terms"] = technology_source.apply(
    lambda x: extract_terms(x, ai_terms)
)

clean_df["High Value Cloud Terms"] = technology_source.apply(
    lambda x: extract_terms(x, cloud_terms)
)

clean_df["High Value Platform Terms"] = technology_source.apply(
    lambda x: extract_terms(x, platform_terms)
)

clean_df["High Value Dev Terms"] = technology_source.apply(
    lambda x: extract_terms(x, dev_terms)
)


# Count actual evidence
clean_df["High Value Technology Count"] = (
    clean_df["High Value AI Terms"].apply(len)
    + clean_df["High Value Cloud Terms"].apply(len)
    + clean_df["High Value Platform Terms"].apply(len)
    + clean_df["High Value Dev Terms"].apply(len)
)

print("Controlled technology extraction completed.")
print()
print(
    clean_df["High Value Technology Count"]
    .describe()
)

print("\nDistribution:")
print(
    clean_df["High Value Technology Count"]
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# FINAL PRODUCTION VALIDATION
# ============================================================

print("=" * 60)
print("FINAL PRODUCTION VALIDATION")
print("=" * 60)

# 1. Dataset size
print("\nDataset:")
print("Rows:", len(final_verification))
print("Columns:", len(final_verification.columns))

# 2. Required scoring columns
required_columns = [
    "Technology Fit Score V2",
    "Company Fit Score",
    "Growth Intent Score",
    "Funding Expansion Score",
    "Marketing Readiness Score",
    "Data Confidence Score",
    "Prospect Score V2",
    "Lead Priority V2",
    "Prospect Rank V2"
]

print("\nRequired scoring columns:")
missing_required = []

for col in required_columns:
    if col not in final_verification.columns:
        missing_required.append(col)
        print("MISSING:", col)
    else:
        print("OK:", col)

# 3. Missing scoring values
print("\nMissing scoring values:")

for col in required_columns:
    if col in final_verification.columns:
        print(
            col,
            "->",
            final_verification[col].isna().sum()
        )

# 4. Score bounds
score_bounds_check = (
    (final_verification["Prospect Score V2"] >= 0) &
    (final_verification["Prospect Score V2"] <= 100)
)

print(
    "\nProspect scores outside 0-100:",
    (~score_bounds_check).sum()
)

# 5. Technology zero-evidence validation
zero_evidence_errors = (
    (final_verification["Technology Evidence Count"] == 0) &
    (final_verification["Technology Fit Score V2"] != 0)
).sum()

print(
    "Zero-evidence records with non-zero Technology Fit V2:",
    zero_evidence_errors
)

# 6. Lead priority consistency
def expected_priority(score):
    if score >= 90:
        return "Tier 1 - High Priority"
    elif score >= 75:
        return "Tier 2 - Priority"
    elif score >= 50:
        return "Tier 3 - Nurture"
    else:
        return "Tier 4 - Low Priority"

expected_priorities = (
    final_verification["Prospect Score V2"]
    .apply(expected_priority)
)

priority_errors = (
    expected_priorities != final_verification["Lead Priority V2"]
).sum()

print(
    "Lead Priority V2 inconsistencies:",
    priority_errors
)

# 7. Company name missing
print(
    "Missing Company Names:",
    final_verification["Company Name"].isna().sum()
)

# 8. Exact duplicate check using scalar columns
list_columns = [
    col for col in final_verification.columns
    if final_verification[col].apply(
        lambda x: isinstance(x, list)
    ).any()
]

scalar_columns = [
    col for col in final_verification.columns
    if col not in list_columns
]

duplicate_rows = final_verification.duplicated(
    subset=scalar_columns
).sum()

print(
    "Duplicate rows based on scalar columns:",
    duplicate_rows
)

# 9. Final status
validation_passed = (
    len(final_verification) == 313
    and len(missing_required) == 0
    and all(
        final_verification[col].isna().sum() == 0
        for col in required_columns
    )
    and (~score_bounds_check).sum() == 0
    and zero_evidence_errors == 0
    and priority_errors == 0
    and final_verification["Company Name"].isna().sum() == 0
    and duplicate_rows == 0
)

print("\n" + "=" * 60)

if validation_passed:
    print("FINAL VALIDATION: PASSED")
else:
    print("FINAL VALIDATION: FAILED")

print("=" * 60)
```
