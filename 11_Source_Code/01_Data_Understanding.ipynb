```python
output_file = "02_Data_Quality/Day_1_Data_Quality_Audit.xlsx"

# Missing value report
missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage": (
        df.isna().mean().values * 100
    ).round(2),
    "Non_Missing_Count": df.notna().sum().values
})

missing_report = missing_report.sort_values(
    "Missing_Percentage",
    ascending=False
)

# Exact duplicate rows
duplicate_rows = df[df.duplicated(keep=False)].copy()

# Exact locations of missing cells
missing_cells = []

for row_index, row in df.iterrows():
    for column in df.columns:
        if pd.isna(row[column]):
            missing_cells.append({
                "Excel_Row": row_index + 2,
                "Column": column,
                "Column_Number": df.columns.get_loc(column) + 1
            })

missing_cell_map = pd.DataFrame(missing_cells)

# Save everything to Excel
with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    df.to_excel(
        writer,
        sheet_name="Raw_Data",
        index=False
    )

    missing_report.to_excel(
        writer,
        sheet_name="Missing_Value_Report",
        index=False
    )

    missing_cell_map.to_excel(
        writer,
        sheet_name="Missing_Cell_Map",
        index=False
    )

    duplicate_rows.to_excel(
        writer,
        sheet_name="Duplicate_Report",
        index=False
    )

print("Day 1 Data Quality Audit created:")
print(output_file)
```

```python
field_rules = {

    "Company": [
        "Company Name",
        "Website",
        "Company Linkedin Url",
        "Company Country",
        "Company City",
        "Company State",
        "Company Address",
        "Industry",
        "# Employees",
        "Annual Revenue",
        "Founded Year",
        "Facebook Url",
        "Keywords",
        "Technology Areas"
    ],

    "Contact": [
        "Contact"
    ],

    "Technology Signals": [
        "AI/ML Signals",
        "API/Platform Signals",
        "Cloud/Data Signals",
        "Software Development Signals"
    ],

    "Business Signals": [
        "Hiring/Growth Signals",
        "Funding/Expansion Signals",
        "Total Funding",
        "Latest Funding",
        "Latest Funding Amount",
        "Last Raised At"
    ],

    "Marketing": [
        "Marketing Segment",
        "Marketing Readiness"
    ],

    "Data Quality": [
        "Data Quality Status"
    ],

    "Source": [
        "Source_File",
        "Apollo Account Id",
        "Apollo Record Id"
    ]
}

classification_rows = []

for col in df.columns:

    category = "Other"

    for cat, fields in field_rules.items():
        if col in fields:
            category = cat
            break

    # Determine field type
    if col in [
        "Marketing Segment",
        "Marketing Readiness",
        "Data Quality Status",
        "AI/ML Signals",
        "API/Platform Signals",
        "Cloud/Data Signals",
        "Software Development Signals",
        "Hiring/Growth Signals",
        "Funding/Expansion Signals"
    ]:
        field_type = "Derived"

    elif col in [
        "Primary NexziByte Service",
        "Secondary NexziByte Service"
    ]:
        field_type = "Derived"

    else:
        field_type = "Raw"

    # Required fields
    required_fields = [
        "Company Name",
        "Website",
        "Industry",
        "# Employees",
        "Company Country"
    ]

    required = "Yes" if col in required_fields else "No"

    # Action
    action = "Keep"

    classification_rows.append({
        "Field_Name": col,
        "Category": category,
        "Field_Type": field_type,
        "Required": required,
        "Action": action,
        "Notes": ""
    })

field_classification = pd.DataFrame(classification_rows)

field_classification
```
